```python
# ========================= TADP_VR_SYSTEMATIC_SENSITIVITY_ANALYSIS.py ===============
# Trustworthy AI Data Preparation (TADP) — Systematic Sensitivity Analysis
# ========================================================================

import os, io, glob, time, uuid, shutil, json, hashlib, warnings, math, random, threading, csv
from datetime import datetime, timezone
from typing import Dict, Any, List, Tuple

import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
import seaborn as sns
from IPython.display import display

warnings.filterwarnings("ignore")

# -------- Matplotlib appearance --------
mpl.rcParams.update({
    "figure.dpi": 600,
    "savefig.dpi": 600,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "grid.alpha": 0.25,
    "legend.frameon": False,
    "font.size": 9,
    "figure.autolayout": True,
    "text.usetex": False,
    "font.family": "DejaVu Sans",
    "font.sans-serif": ["DejaVu Sans", "Liberation Sans"]
})

# Professional Q1 color scheme
COLORS = {
    'blue1': '#1f4e79',    # Dark blue
    'blue2': '#2e75b6',    # Medium blue
    'blue3': '#9dc3e6',    # Light blue
    'orange1': '#ed7d31',  # Orange
    'gray1': '#d0cece',    # Light gray
    'gray2': '#7f7f7f',    # Medium gray
    'gray3': '#2f2f2f',    # Dark gray
}

import psutil
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, MaxAbsScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score, accuracy_score, precision_score, recall_score, f1_score

# ---------------- Global config ----------------
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

# Configuration
RANDOM_STATE     = 42
N_RUNS           = 1
NUM_CLIENTS      = 10
NUM_ROUNDS_FL    = 4
LOCAL_EPOCHS     = 1
EPOCHS_CENTRAL   = 2
BATCH_SIZE       = 64
CLIENT_FRACTION  = 0.6

# TADP thresholds & P-score weights - BASELINE
GOOD_CUT = 3.0
HIGH_CUT = 3.5
WEIGHTS_PSCORE_DEFAULT = {"dim1":0.25,"dim2":0.15,"dim3":0.10,"dim4":0.10,"dim5":0.30,"dim6":0.10}

DATA_FILENAME  = "diabetes_130US.csv"
DATA_CACHE_DIR = "./data_cache"
os.makedirs(DATA_CACHE_DIR, exist_ok=True)

# File paths
TABLES_DIR = "./tables"
FIG_DIR    = os.path.join(TABLES_DIR, "figs")
for p in [TABLES_DIR, FIG_DIR]: os.makedirs(p, exist_ok=True)

# ---------------- Systematic Sensitivity Scenarios ----------------
def generate_weight_scenarios():
    """Generate 6 weight scenarios with sum = 1"""
    scenarios = []

    # Scenario 1: Balanced weights
    scenarios.append({
        "name": "Weight_Balanced",
        "weights": {"dim1":0.17,"dim2":0.17,"dim3":0.16,"dim4":0.16,"dim5":0.17,"dim6":0.17},
        "good_cut": GOOD_CUT,
        "high_cut": HIGH_CUT,
        "review_accept_ratio": 0.6,
        "type": "weight_variation"
    })

    # Scenario 2: Focus on Source Reliability and Data Quality
    scenarios.append({
        "name": "Weight_SourceData_Focus",
        "weights": {"dim1":0.30,"dim2":0.25,"dim3":0.10,"dim4":0.10,"dim5":0.15,"dim6":0.10},
        "good_cut": GOOD_CUT,
        "high_cut": HIGH_CUT,
        "review_accept_ratio": 0.6,
        "type": "weight_variation"
    })

    # Scenario 3: Focus on Regulatory Compliance
    scenarios.append({
        "name": "Weight_Regulatory_Focus",
        "weights": {"dim1":0.15,"dim2":0.15,"dim3":0.10,"dim4":0.10,"dim5":0.40,"dim6":0.10},
        "good_cut": GOOD_CUT,
        "high_cut": HIGH_CUT,
        "review_accept_ratio": 0.6,
        "type": "weight_variation"
    })

    # Scenario 4: Mixed emphasis
    scenarios.append({
        "name": "Weight_Mixed_Emphasis",
        "weights": {"dim1":0.20,"dim2":0.20,"dim3":0.15,"dim4":0.15,"dim5":0.20,"dim6":0.10},
        "good_cut": GOOD_CUT,
        "high_cut": HIGH_CUT,
        "review_accept_ratio": 0.6,
        "type": "weight_variation"
    })

    # Scenario 5: Extreme Reliability Focus
    scenarios.append({
        "name": "Weight_Extreme_Reliability",
        "weights": {"dim1":0.40,"dim2":0.20,"dim3":0.10,"dim4":0.10,"dim5":0.10,"dim6":0.10},
        "good_cut": GOOD_CUT,
        "high_cut": HIGH_CUT,
        "review_accept_ratio": 0.6,
        "type": "weight_variation"
    })

    # Scenario 6: Quality and Timeliness Focus
    scenarios.append({
        "name": "Weight_Quality_Timeliness",
        "weights": {"dim1":0.15,"dim2":0.30,"dim3":0.10,"dim4":0.10,"dim5":0.15,"dim6":0.20},
        "good_cut": GOOD_CUT,
        "high_cut": HIGH_CUT,
        "review_accept_ratio": 0.6,
        "type": "weight_variation"
    })

    return scenarios

def generate_threshold_scenarios():
    """Generate 6 threshold scenarios with more variation"""
    scenarios = []

    # Scenario 1: Very Lenient thresholds
    scenarios.append({
        "name": "Threshold_Very_Lenient",
        "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
        "good_cut": 2.0,
        "high_cut": 2.5,
        "review_accept_ratio": 0.8,
        "type": "threshold_variation"
    })

    # Scenario 2: Lenient thresholds
    scenarios.append({
        "name": "Threshold_Lenient",
        "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
        "good_cut": 2.5,
        "high_cut": 3.0,
        "review_accept_ratio": 0.6,
        "type": "threshold_variation"
    })

    # Scenario 3: Moderate thresholds
    scenarios.append({
        "name": "Threshold_Moderate",
        "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
        "good_cut": 2.8,
        "high_cut": 3.3,
        "review_accept_ratio": 0.6,
        "type": "threshold_variation"
    })

    # Scenario 4: Strict thresholds
    scenarios.append({
        "name": "Threshold_Strict",
        "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
        "good_cut": 3.2,
        "high_cut": 3.8,
        "review_accept_ratio": 0.6,
        "type": "threshold_variation"
    })

    # Scenario 5: Very Strict thresholds
    scenarios.append({
        "name": "Threshold_Very_Strict",
        "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
        "good_cut": 3.5,
        "high_cut": 4.2,
        "review_accept_ratio": 0.4,
        "type": "threshold_variation"
    })

    # Scenario 6: Mixed thresholds
    scenarios.append({
        "name": "Threshold_Mixed",
        "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
        "good_cut": 2.7,
        "high_cut": 3.7,
        "review_accept_ratio": 0.7,
        "type": "threshold_variation"
    })

    return scenarios

def generate_combined_scenarios():
    """Generate 6 combined weight + threshold scenarios"""
    scenarios = []

    # Scenario 1: Balanced weights + Very Lenient thresholds
    scenarios.append({
        "name": "Combined_Balanced_Very_Lenient",
        "weights": {"dim1":0.17,"dim2":0.17,"dim3":0.16,"dim4":0.16,"dim5":0.17,"dim6":0.17},
        "good_cut": 2.0,
        "high_cut": 2.5,
        "review_accept_ratio": 0.8,
        "type": "combined_variation"
    })

    # Scenario 2: SourceData focus + Moderate thresholds
    scenarios.append({
        "name": "Combined_SourceData_Moderate",
        "weights": {"dim1":0.30,"dim2":0.25,"dim3":0.10,"dim4":0.10,"dim5":0.15,"dim6":0.10},
        "good_cut": 2.8,
        "high_cut": 3.3,
        "review_accept_ratio": 0.6,
        "type": "combined_variation"
    })

    # Scenario 3: Regulatory focus + Strict thresholds
    scenarios.append({
        "name": "Combined_Regulatory_Strict",
        "weights": {"dim1":0.15,"dim2":0.15,"dim3":0.10,"dim4":0.10,"dim5":0.40,"dim6":0.10},
        "good_cut": 3.2,
        "high_cut": 3.8,
        "review_accept_ratio": 0.6,
        "type": "combined_variation"
    })

    # Scenario 4: Mixed emphasis + Very Strict thresholds
    scenarios.append({
        "name": "Combined_Mixed_Very_Strict",
        "weights": {"dim1":0.20,"dim2":0.20,"dim3":0.15,"dim4":0.15,"dim5":0.20,"dim6":0.10},
        "good_cut": 3.5,
        "high_cut": 4.2,
        "review_accept_ratio": 0.4,
        "type": "combined_variation"
    })

    # Scenario 5: Extreme Reliability + Mixed thresholds
    scenarios.append({
        "name": "Combined_ExtremeReliability_Mixed",
        "weights": {"dim1":0.40,"dim2":0.20,"dim3":0.10,"dim4":0.10,"dim5":0.10,"dim6":0.10},
        "good_cut": 2.7,
        "high_cut": 3.7,
        "review_accept_ratio": 0.7,
        "type": "combined_variation"
    })

    # Scenario 6: Quality Focus + Lenient thresholds
    scenarios.append({
        "name": "Combined_Quality_Lenient",
        "weights": {"dim1":0.15,"dim2":0.30,"dim3":0.10,"dim4":0.10,"dim5":0.15,"dim6":0.20},
        "good_cut": 2.5,
        "high_cut": 3.0,
        "review_accept_ratio": 0.6,
        "type": "combined_variation"
    })

    return scenarios

# Generate all sensitivity cases
BASELINE_CASE = {
    "name": "Baseline",
    "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
    "good_cut": GOOD_CUT,
    "high_cut": HIGH_CUT,
    "review_accept_ratio": 0.6,
    "type": "baseline"
}

SENSITIVITY_CASES = [BASELINE_CASE] + generate_weight_scenarios() + generate_threshold_scenarios() + generate_combined_scenarios()

# ---------------- Core Functions ----------------
def preprocess_diabetes(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df.columns = df.columns.str.lower().str.replace(r'[^a-z0-9]+', '_', regex=True)
    df.replace('?', np.nan, inplace=True)

    if 'age' in df.columns and df['age'].dtype == object:
        df['age'] = df['age'].str.extract(r'(\d+)').astype(float)

    if 'readmitted' in df.columns:
        mapping = {'NO': 0, '<30': 1, '>30': 2, 'NONE': 0}
        s = df['readmitted'].astype(str).str.strip().str.upper()
        df['readmitted'] = s.map(mapping).fillna(0).astype('int8')

    df.drop(columns=['encounter_id', 'patient_nbr'], errors='ignore', inplace=True)
    return df

def build_preproc(df: pd.DataFrame, target: str = "readmitted"):
    from sklearn.preprocessing import StandardScaler
    from sklearn.compose import ColumnTransformer
    X = df.drop(columns=[target], errors="ignore").copy()
    numeric_cols = X.select_dtypes(include=['number']).columns.tolist()
    pre = ColumnTransformer(transformers=[("num", StandardScaler(), numeric_cols)], remainder="drop", sparse_threshold=0.0)
    return pre

def build_model(input_dim):
    inp = keras.Input(shape=(input_dim,), dtype=tf.float32)
    x = layers.Dense(128, activation="relu")(inp)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dropout(0.2)(x)
    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dropout(0.2)(x)
    x = layers.Dense(64, activation="relu")(x)
    out = layers.Dense(3, activation="softmax", dtype=tf.float32)(x)
    model = keras.Model(inp, out)
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=0.001), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    return model

def eval_metrics(model, Xte, yte):
    p = model.predict(Xte, verbose=0)
    yhat = np.argmax(p, axis=1)
    precision = precision_score(yte, yhat, average='macro', zero_division=0)
    recall = recall_score(yte, yhat, average='macro', zero_division=0)
    f1 = f1_score(yte, yhat, average='macro', zero_division=0)

    try:
        n_classes = len(np.unique(yte))
        if n_classes == 2:
            auc = roc_auc_score(yte, p[:, 1])
        elif n_classes == 3:
            try:
                auc = roc_auc_score(yte, p, multi_class='ovo', average='macro')
            except:
                auc = roc_auc_score(yte, p, multi_class='ovr', average='macro')
        else:
            auc = np.nan
    except:
        auc = np.nan

    return {"roc_auc": float(auc), "accuracy": float(accuracy_score(yte, yhat)), "precision": float(precision), "recall": float(recall), "f1": float(f1)}

def get_wvec(m):
    vec=[]
    for w in m.get_weights(): vec.append(w.flatten())
    return np.concatenate(vec) if vec else np.array([], np.float32)

def set_wvec(m, vec):
    shapes=[w.shape for w in m.get_weights()]
    out=[]; i=0
    for s in shapes:
        n=int(np.prod(s)); out.append(vec[i:i+n].reshape(s)); i+=n
    m.set_weights(out)

def split_clients(df: pd.DataFrame, num_clients=NUM_CLIENTS, target_col="readmitted", alpha=0.5):
    df = df.sample(frac=1, random_state=RANDOM_STATE).reset_index(drop=True)
    labels = df[target_col].values if target_col in df.columns else np.zeros(len(df), int)
    client_ids = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")[:num_clients]
    buckets = {cid: [] for cid in client_ids}
    rng = np.random.default_rng(RANDOM_STATE + 77)

    for y in np.unique(labels):
        idx = np.where(labels == y)[0]
        rng.shuffle(idx)
        props = rng.dirichlet([alpha] * num_clients)
        cuts = (np.cumsum(props) * len(idx)).astype(int)
        parts = np.split(idx, cuts[:-1])
        for cid, part in zip(client_ids, parts):
            buckets[cid].extend(part.tolist())

    client_data = {cid: df.loc[sorted(buckets[cid])].copy() for cid in client_ids}
    feature_num_cols = df.drop(columns=[target_col], errors="ignore").select_dtypes(include=np.number).columns.tolist()

    for j, cid in enumerate(client_ids):
        if not feature_num_cols: break
        shift = (j - (num_clients - 1) / 2.0) * 0.25
        noise = rng.normal(0, 0.05, size=len(feature_num_cols))
        client_data[cid].loc[:, feature_num_cols] = client_data[cid][feature_num_cols] + shift + noise

        if target_col in client_data[cid].columns:
            y = pd.to_numeric(client_data[cid][target_col], errors="coerce").fillna(0).astype(int)
            y = np.clip(y, 0, 2)
            client_data[cid][target_col] = y.astype(np.int8)

    return client_ids, client_data

def prepare_xy_without_reprocessing(clients: Dict[str, pd.DataFrame], target="readmitted", test_size=0.2):
    def can_stratify(y: np.ndarray, test_size: float) -> bool:
        vals, counts = np.unique(y, return_counts=True)
        if len(vals) < 2: return False
        for n in counts:
            if n < 2 or n * test_size < 1 or n * (1.0 - test_size) < 1: return False
        return True

    def sanitize_labels(y_like) -> np.ndarray:
        y = pd.to_numeric(pd.Series(y_like), errors="coerce").fillna(0).astype(int).values
        return np.clip(y, 0, 2).astype(np.int32)

    prepped = clients
    splits = {}
    train_parts, test_parts = [], []

    for cid, df in prepped.items():
        if target not in df.columns or len(df) == 0: continue
        df = df.copy()
        df[target] = sanitize_labels(df[target].values)
        X = df.drop(columns=[target])
        y = df[target].values
        strat = y if can_stratify(y, test_size) else None

        Xtr_raw, Xte_raw, ytr_raw, yte_raw = train_test_split(X, y, test_size=test_size, random_state=RANDOM_STATE, stratify=strat)
        ytr_raw = sanitize_labels(ytr_raw)
        yte_raw = sanitize_labels(yte_raw)
        splits[cid] = (Xtr_raw, Xte_raw, ytr_raw, yte_raw)
        train_parts.append(pd.concat([Xtr_raw.reset_index(drop=True), pd.Series(ytr_raw, name=target)], axis=1))
        test_parts.append(pd.concat([Xte_raw.reset_index(drop=True), pd.Series(yte_raw, name=target)], axis=1))

    if not train_parts: return np.array([]), np.array([]), np.array([]), np.array([]), {}

    train_all_df = pd.concat(train_parts, ignore_index=True)
    pre = build_preproc(train_all_df, target)
    pre.fit(train_all_df.drop(columns=[target], errors='ignore'))

    client_arrays = {}
    all_Xtr, all_ytr, all_Xte, all_yte = [], [], [], []
    for cid, (Xtr_raw, Xte_raw, ytr_raw, yte_raw) in splits.items():
        Xtr = pre.transform(Xtr_raw)
        Xte = pre.transform(Xte_raw)
        Xtr = Xtr.toarray() if hasattr(Xtr, "toarray") else np.asarray(Xtr, dtype=np.float32)
        Xte = Xte.toarray() if hasattr(Xte, "toarray") else np.asarray(Xte, dtype=np.float32)
        Xtr = np.nan_to_num(Xtr).astype(np.float32)
        Xte = np.nan_to_num(Xte).astype(np.float32)
        ytr = sanitize_labels(ytr_raw)
        yte = sanitize_labels(yte_raw)
        client_arrays[cid] = (Xtr, ytr)
        all_Xtr.append(Xtr); all_ytr.append(ytr)
        all_Xte.append(Xte); all_yte.append(yte)

    Xtr_all = np.vstack(all_Xtr) if all_Xtr else np.array([], dtype=np.float32)
    ytr_all = np.concatenate(all_ytr) if all_ytr else np.array([], dtype=np.int32)
    Xte_global = np.vstack(all_Xte) if all_Xte else np.array([], dtype=np.float32)
    yte_global = np.concatenate(all_yte) if all_yte else np.array([], dtype=np.int32)

    if yte_global.size and len(np.unique(yte_global)) < 2:
        all_df = pd.concat([df for df in prepped.values()], ignore_index=True)
        all_df = all_df.copy()
        all_df[target] = sanitize_labels(all_df[target].values)
        X_all = all_df.drop(columns=[target], errors='ignore')
        y_all = all_df[target].values
        strat_all = y_all if can_stratify(y_all, test_size) else None
        _, Xte_raw_fb, _, yte_raw_fb = train_test_split(X_all, y_all, test_size=test_size, random_state=RANDOM_STATE, stratify=strat_all)
        Xte_fb = pre.transform(Xte_raw_fb)
        Xte_fb = Xte_fb.toarray() if hasattr(Xte_fb, "toarray") else np.asarray(Xte_fb, dtype=np.float32)
        Xte_global = np.nan_to_num(Xte_fb).astype(np.float32)
        yte_global = sanitize_labels(yte_raw_fb)

    return Xtr_all, Xte_global, ytr_all, yte_global, client_arrays

def load_dataset():
    global _DATASET_CACHE
    cache_path = os.path.join(DATA_CACHE_DIR, DATA_FILENAME)
    if _DATASET_CACHE is not None: return _DATASET_CACHE
    if os.path.exists(cache_path):
        _DATASET_CACHE = pd.read_csv(cache_path); return _DATASET_CACHE
    if os.path.exists(DATA_FILENAME):
        df = pd.read_csv(DATA_FILENAME); df.to_csv(cache_path, index=False); _DATASET_CACHE = df; return df

    print("\n⚠️  Dataset not found in cache or working directory.")
    user_path = input("Paste local file path here (or press Enter to cancel): ").strip()
    if not user_path: raise FileNotFoundError("No dataset path provided. Aborting.")
    if not os.path.exists(user_path): raise FileNotFoundError(f"Provided path does not exist: {user_path}")
    shutil.copy2(user_path, cache_path)
    _DATASET_CACHE = pd.read_csv(cache_path)
    return _DATASET_CACHE

# ---------------- TADP-VR Core Functions ----------------
def dims_for_client_realistic(client_id: str, total_clients=NUM_CLIENTS):
    """Generate MUCH more diverse P-scores to ensure sensitivity"""
    client_num = ord(client_id[0]) - ord('A') if client_id and client_id[0].isalpha() else 0
    client_position = client_num / max(1, total_clients - 1)

    # Create MUCH wider variation in base scores
    if client_position < 0.2:  # Bottom 20% - Low quality clients
        base_scores = [1.2, 1.5, 1.8, 1.4, 1.3, 1.6]  # Much lower scores
    elif client_position < 0.5:  # Next 30% - Medium-low quality
        base_scores = [2.2, 2.5, 2.3, 2.1, 2.4, 2.6]
    elif client_position < 0.8:  # Next 30% - Medium-high quality
        base_scores = [3.2, 3.5, 3.3, 3.1, 3.4, 3.6]
    else:  # Top 20% - High quality clients
        base_scores = [4.2, 4.5, 4.3, 4.1, 4.4, 4.6]  # Much higher scores

    random.seed(hash(client_id) % 1000)
    final_scores = []
    for base_score in base_scores:
        # Increase variation range significantly
        variation = random.uniform(-0.4, 0.4)  # Increased from ±0.15 to ±0.4
        final_score = max(0.5, min(5.0, base_score + variation))  # Wider bounds
        final_scores.append(round(final_score, 2))

    return {"dim1": final_scores[0], "dim2": final_scores[1], "dim3": final_scores[2],
            "dim4": final_scores[3], "dim5": final_scores[4], "dim6": final_scores[5]}

def calculate_pscore(dims: Dict[str,float], weights: Dict[str,float]):
    pscore = 0.0
    for dim, weight in weights.items():
        pscore += weight * dims[dim]
    return pscore

def tadp_vr_partition(client_ids: List[str], scores: Dict[str,float], good_cut, high_cut, review_accept_ratio=0.6):
    sorted_clients = sorted(client_ids, key=lambda cid: scores[cid])
    auto_reject = [cid for cid in sorted_clients if scores[cid] < good_cut]
    review = [cid for cid in sorted_clients if good_cut <= scores[cid] < high_cut]
    auto_accept = [cid for cid in sorted_clients if scores[cid] >= high_cut]

    review_sorted = sorted(review, key=lambda cid: scores[cid])
    review_accept_count = max(0, int(len(review) * review_accept_ratio))
    review_accept = review_sorted[-review_accept_count:] if review_sorted else []
    review_reject = review_sorted[:-review_accept_count] if review_sorted else []

    accepted = auto_accept + review_accept
    rejected = auto_reject + review_reject

    return accepted, rejected

def fed_round_sensitivity(global_model, round_idx, client_arrays, client_ids, sensitivity_case):
    import gc
    gc.collect()
    input_dim = global_model.input_shape[-1]

    dimension_scores = {cid: dims_for_client_realistic(cid, len(client_ids)) for cid in client_ids}
    scores = {cid: calculate_pscore(dimension_scores[cid], sensitivity_case["weights"]) for cid in client_ids}

    accepted, rejected = tadp_vr_partition(client_ids, scores, sensitivity_case["good_cut"], sensitivity_case["high_cut"], sensitivity_case["review_accept_ratio"])

    accepted = [cid for cid in accepted if cid in client_arrays]
    if not accepted:
        accepted = list(client_arrays.keys())[:2]
        print(f"    ⚠️  No clients accepted, using fallback: {accepted}")

    client_weights, client_n = [], []
    for cid in accepted:
        Xc, yc = client_arrays[cid]
        if len(Xc) == 0: continue
        cm = build_model(input_dim)
        set_wvec(cm, get_wvec(global_model))
        cm.fit(Xc, yc, epochs=LOCAL_EPOCHS, batch_size=BATCH_SIZE, verbose=0)
        client_weights.append(get_wvec(cm))
        client_n.append(len(Xc))

    if not client_weights:
        print("    ⚠️  No client updates received, keeping global model")
        return global_model, len(accepted), accepted, rejected, scores

    avg_weight = np.average(client_weights, axis=0, weights=client_n)
    set_wvec(global_model, avg_weight)

    return global_model, len(accepted), accepted, rejected, scores

def run_federated_sensitivity(client_ids, client_arrays, Xte, yte, sensitivity_case):
    print(f"\n🔬 Running Federated Sensitivity: {sensitivity_case['name']}")
    print(f"   Weights: {sensitivity_case['weights']}")
    print(f"   Thresholds: good_cut={sensitivity_case['good_cut']}, high_cut={sensitivity_case['high_cut']}")

    input_dim = Xte.shape[1]
    global_model = build_model(input_dim)

    start_time = time.time()
    total_clients_trained = 0
    round_clients = []
    all_accepted_clients = []
    all_rejected_clients = []
    all_scores = {}

    for round_idx in range(NUM_ROUNDS_FL):
        global_model, round_clients_count, accepted, rejected, scores = fed_round_sensitivity(
            global_model, round_idx, client_arrays, client_ids, sensitivity_case
        )
        total_clients_trained += round_clients_count
        round_clients.append(round_clients_count)
        all_accepted_clients = accepted
        all_rejected_clients = rejected
        all_scores = scores

    elapsed = time.time() - start_time
    metrics = eval_metrics(global_model, Xte, yte)
    acceptance_rate = len(all_accepted_clients) / len(client_ids) * 100

    return {
        "metrics": metrics,
        "elapsed": elapsed,
        "avg_clients_per_round": np.mean(round_clients),
        "total_clients_trained": total_clients_trained,
        "accepted_clients": all_accepted_clients,
        "rejected_clients": all_rejected_clients,
        "acceptance_rate": acceptance_rate,
        "client_scores": all_scores,
    }

def run_sensitivity_analysis(client_ids, raw_clients, client_arrays, Xte, yte):
    print("\n" + "="*80)
    print("🔬 TADP-VR SYSTEMATIC SENSITIVITY ANALYSIS")
    print("="*80)

    results = []
    baseline_result = None

    for case_idx, case in enumerate(SENSITIVITY_CASES):
        print(f"\n🧪 Sensitivity Case {case_idx+1}/{len(SENSITIVITY_CASES)}: {case['name']}")
        print(f"   Type: {case['type']}")
        print(f"   Weights: {case['weights']}")
        print(f"   Thresholds: good_cut={case['good_cut']}, high_cut={case['high_cut']}")

        try:
            case_result = run_federated_sensitivity(client_ids, client_arrays, Xte, yte, case)

            if case["name"] == "Baseline":
                baseline_result = case_result

            # Calculate average P-score for accepted clients
            if case_result["accepted_clients"]:
                avg_pscore = np.mean([case_result["client_scores"][cid] for cid in case_result["accepted_clients"]])
            else:
                avg_pscore = 0.0

            if baseline_result is not None and case["name"] != "Baseline":
                delta_accuracy = case_result["metrics"]["accuracy"] - baseline_result["metrics"]["accuracy"]
                delta_f1 = case_result["metrics"]["f1"] - baseline_result["metrics"]["f1"]
                delta_roc_auc = case_result["metrics"]["roc_auc"] - baseline_result["metrics"]["roc_auc"]
                delta_precision = case_result["metrics"]["precision"] - baseline_result["metrics"]["precision"]
                delta_recall = case_result["metrics"]["recall"] - baseline_result["metrics"]["recall"]

                baseline_avg_pscore = np.mean([baseline_result["client_scores"][cid] for cid in baseline_result["accepted_clients"]]) if baseline_result["accepted_clients"] else 0
                delta_pscore = avg_pscore - baseline_avg_pscore

                accepted_set = set(case_result["accepted_clients"])
                baseline_set = set(baseline_result["accepted_clients"])
                clients_gained = list(accepted_set - baseline_set)
                clients_lost = list(baseline_set - accepted_set)

                delta_acceptance_rate = case_result["acceptance_rate"] - baseline_result["acceptance_rate"]
            else:
                delta_accuracy = 0.0
                delta_f1 = 0.0
                delta_roc_auc = 0.0
                delta_precision = 0.0
                delta_recall = 0.0
                delta_pscore = 0.0
                delta_acceptance_rate = 0.0
                clients_gained = []
                clients_lost = []

            results.append({
                "case_name": case["name"],
                "case_type": case["type"],
                "weights": case["weights"],
                "good_cut": case["good_cut"],
                "high_cut": case["high_cut"],
                "accuracy": case_result["metrics"]["accuracy"],
                "f1": case_result["metrics"]["f1"],
                "roc_auc": case_result["metrics"]["roc_auc"],
                "precision": case_result["metrics"]["precision"],
                "recall": case_result["metrics"]["recall"],
                "acceptance_rate": case_result["acceptance_rate"],
                "avg_pscore": avg_pscore,
                "total_clients_trained": case_result["total_clients_trained"],
                "elapsed_time": case_result["elapsed"],
                "delta_accuracy": delta_accuracy,
                "delta_f1": delta_f1,
                "delta_roc_auc": delta_roc_auc,
                "delta_precision": delta_precision,
                "delta_recall": delta_recall,
                "delta_pscore": delta_pscore,
                "delta_acceptance_rate": delta_acceptance_rate,
                "accepted_clients": case_result["accepted_clients"],
                "clients_gained": clients_gained,
                "clients_lost": clients_lost,
                "num_clients_gained": len(clients_gained),
                "num_clients_lost": len(clients_lost),
            })

            print(f"✅ Completed: {case['name']}")

        except Exception as e:
            print(f"❌ Failed {case['name']}: {e}")
            import traceback
            traceback.print_exc()

    return results, baseline_result

# ---------------- REFINED TABLES FOR Q1 ARTICLE ----------------
def create_configuration_table(results):
    """Table 1: Scenario configuration and performance deltas"""
    print("\n📊 Creating Configuration and Performance Table...")

    df = pd.DataFrame(results)

    table_data = []
    for _, row in df.iterrows():
        # Format weights for better readability
        weights_str = ", ".join([f"{k}:{v:.2f}" for k, v in row['weights'].items()])

        if row['case_type'] == 'baseline':
            table_data.append({
                'Scenario': row['case_name'],
                'Type': row['case_type'].title(),
                'good_cut': row['good_cut'],
                'high_cut': row['high_cut'],
                'Weights': weights_str,
                'Accuracy': f"{row['accuracy']*100:.2f}%",
                'F1-Score': f"{row['f1']:.4f}",
                'ROC-AUC': f"{row['roc_auc']:.4f}",
                'Acceptance Rate': f"{row['acceptance_rate']:.1f}%",
                'Avg P-score': f"{row['avg_pscore']:.3f}",
                'Time (s)': f"{row['elapsed_time']:.2f}"
            })
        else:
            table_data.append({
                'Scenario': row['case_name'],
                'Type': row['case_type'].replace('_', ' ').title(),
                'good_cut': row['good_cut'],
                'high_cut': row['high_cut'],
                'Weights': weights_str,
                'Accuracy': f"{row['accuracy']*100:.2f}%",
                'Δ Accuracy': f"{row['delta_accuracy']*100:+.2f}%",
                'F1-Score': f"{row['f1']:.4f}",
                'Δ F1-Score': f"{row['delta_f1']:+.4f}",
                'ROC-AUC': f"{row['roc_auc']:.4f}",
                'Δ ROC-AUC': f"{row['delta_roc_auc']:+.4f}",
                'Acceptance Rate': f"{row['acceptance_rate']:.1f}%",
                'Δ Acceptance': f"{row['delta_acceptance_rate']:+.1f}%",
                'Avg P-score': f"{row['avg_pscore']:.3f}",
                'Δ P-score': f"{row['delta_pscore']:+.3f}",
                'Time (s)': f"{row['elapsed_time']:.2f}"
            })

    config_df = pd.DataFrame(table_data)

    print("\n" + "="*120)
    print("📋 TABLE 1: SENSITIVITY ANALYSIS CONFIGURATION AND PERFORMANCE")
    print("="*120)
    display(config_df)

    config_df.to_csv(os.path.join(TABLES_DIR, "sensitivity_configuration_table.csv"), index=False)
    print("✅ Configuration table saved to 'tables/sensitivity_configuration_table.csv'")

    return config_df

def create_client_impact_table(results):
    """Table 2: Client impact analysis"""
    print("\n📊 Creating Client Impact Table...")

    df = pd.DataFrame(results)

    table_data = []
    for _, row in df.iterrows():
        # Calculate accepted clients count safely
        accepted_clients_count = len(row['accepted_clients']) if row['accepted_clients'] is not None else 0

        if row['case_type'] == 'baseline':
            table_data.append({
                'Scenario': row['case_name'],
                'Type': row['case_type'].title(),
                'Accepted Clients': accepted_clients_count,
                'Total Clients': NUM_CLIENTS,
                'Acceptance Rate': f"{row['acceptance_rate']:.1f}%",
                'Avg P-score': f"{row['avg_pscore']:.3f}",
                'Clients Trained': int(row['total_clients_trained'])
            })
        else:
            net_change = row['num_clients_gained'] - row['num_clients_lost']
            table_data.append({
                'Scenario': row['case_name'],
                'Type': row['case_type'].replace('_', ' ').title(),
                'Accepted Clients': accepted_clients_count,
                'Clients Gained': row['num_clients_gained'],
                'Clients Lost': row['num_clients_lost'],
                'Net Client Change': net_change,
                'Acceptance Rate': f"{row['acceptance_rate']:.1f}%",
                'Avg P-score': f"{row['avg_pscore']:.3f}",
                'Δ P-score': f"{row['delta_pscore']:+.3f}",
                'Clients Trained': int(row['total_clients_trained'])
            })

    impact_df = pd.DataFrame(table_data)

    print("\n" + "="*100)
    print("📋 TABLE 2: CLIENT IMPACT ANALYSIS")
    print("="*100)
    display(impact_df)

    impact_df.to_csv(os.path.join(TABLES_DIR, "client_impact_table.csv"), index=False)
    print("✅ Client impact table saved to 'tables/client_impact_table.csv'")

    return impact_df

def create_detailed_weights_table(results):
    """Table 3: Detailed weight configurations"""
    print("\n📊 Creating Detailed Weights Table...")

    df = pd.DataFrame(results)

    table_data = []
    for _, row in df.iterrows():
        table_data.append({
            'Scenario': row['case_name'],
            'Type': row['case_type'].replace('_', ' ').title(),
            'dim1 (Reliability)': f"{row['weights']['dim1']:.2f}",
            'dim2 (Quality)': f"{row['weights']['dim2']:.2f}",
            'dim3 (Completeness)': f"{row['weights']['dim3']:.2f}",
            'dim4 (Consistency)': f"{row['weights']['dim4']:.2f}",
            'dim5 (Compliance)': f"{row['weights']['dim5']:.2f}",
            'dim6 (Timeliness)': f"{row['weights']['dim6']:.2f}",
            'good_cut': row['good_cut'],
            'high_cut': row['high_cut'],
            'Sum Weights': f"{sum(row['weights'].values()):.2f}"
        })

    weights_df = pd.DataFrame(table_data)

    print("\n" + "="*120)
    print("📋 TABLE 3: DETAILED WEIGHT CONFIGURATIONS")
    print("="*120)
    display(weights_df)

    weights_df.to_csv(os.path.join(TABLES_DIR, "detailed_weights_table.csv"), index=False)
    print("✅ Detailed weights table saved to 'tables/detailed_weights_table.csv'")

    return weights_df

def create_performance_summary_table(results):
    """Table 4: Comprehensive performance summary"""
    print("\n📊 Creating Performance Summary Table...")

    df = pd.DataFrame(results)

    table_data = []
    for _, row in df.iterrows():
        if row['case_type'] == 'baseline':
            table_data.append({
                'Scenario': row['case_name'],
                'Type': row['case_type'].title(),
                'Accuracy': f"{row['accuracy']*100:.2f}%",
                'F1-Score': f"{row['f1']:.4f}",
                'ROC-AUC': f"{row['roc_auc']:.4f}",
                'Precision': f"{row['precision']:.4f}",
                'Recall': f"{row['recall']:.4f}",
                'Acceptance Rate': f"{row['acceptance_rate']:.1f}%",
                'Avg P-score': f"{row['avg_pscore']:.3f}",
                'Training Time (s)': f"{row['elapsed_time']:.2f}"
            })
        else:
            table_data.append({
                'Scenario': row['case_name'],
                'Type': row['case_type'].replace('_', ' ').title(),
                'Accuracy': f"{row['accuracy']*100:.2f}%",
                'Δ Accuracy': f"{row['delta_accuracy']*100:+.2f}%",
                'F1-Score': f"{row['f1']:.4f}",
                'Δ F1-Score': f"{row['delta_f1']:+.4f}",
                'ROC-AUC': f"{row['roc_auc']:.4f}",
                'Δ ROC-AUC': f"{row['delta_roc_auc']:+.4f}",
                'Precision': f"{row['precision']:.4f}",
                'Recall': f"{row['recall']:.4f}",
                'Acceptance Rate': f"{row['acceptance_rate']:.1f}%",
                'Δ Acceptance': f"{row['delta_acceptance_rate']:+.1f}%",
                'Avg P-score': f"{row['avg_pscore']:.3f}",
                'Δ P-score': f"{row['delta_pscore']:+.3f}",
                'Training Time (s)': f"{row['elapsed_time']:.2f}"
            })

    summary_df = pd.DataFrame(table_data)

    print("\n" + "="*120)
    print("📋 TABLE 4: COMPREHENSIVE PERFORMANCE SUMMARY")
    print("="*120)
    display(summary_df)

    summary_df.to_csv(os.path.join(TABLES_DIR, "performance_summary_table.csv"), index=False)
    print("✅ Performance summary table saved to 'tables/performance_summary_table.csv'")

    return summary_df

# ---------------- Q1 ARTICLE FIGURES ----------------
def create_performance_comparison_figure(results, baseline_result):
    """Main performance comparison figure for Q1 article"""
    print("\n📈 Creating Performance Comparison Figure...")

    df = pd.DataFrame(results)
    non_baseline = df[df['case_name'] != 'Baseline'].copy()

    if len(non_baseline) == 0:
        print("⚠️ No non-baseline results to plot")
        return None

    # Create professional figure with separate subplots
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    ax1, ax2, ax3, ax4 = axes[0, 0], axes[0, 1], axes[1, 0], axes[1, 1]

    # Plot 1: Accuracy comparison
    scenario_names = [row['case_name'].replace('Weight_', 'W_').replace('Threshold_', 'T_').replace('Combined_', 'C_')
                     for _, row in non_baseline.iterrows()]
    accuracy_scores = non_baseline['accuracy'].values * 100

    x_pos = np.arange(len(non_baseline))
    bars = ax1.bar(x_pos, accuracy_scores, color=COLORS['blue2'], alpha=0.8, edgecolor='black')

    # Add baseline line
    baseline_acc = baseline_result["metrics"]["accuracy"] * 100
    ax1.axhline(y=baseline_acc, color=COLORS['blue1'], linestyle='--', linewidth=2,
                label=f'Baseline ({baseline_acc:.1f}%)')

    ax1.set_ylabel('Accuracy (%)', fontweight='bold', fontsize=12)
    ax1.set_title('A. Model Accuracy Across Parameter Configurations', fontweight='bold', fontsize=14)
    ax1.set_xticks(x_pos)
    ax1.set_xticklabels(scenario_names, rotation=45, ha='right', fontsize=9)
    ax1.legend(fontsize=10, loc='upper right')
    ax1.grid(True, alpha=0.3, axis='y')
    ax1.set_ylim([0, max(accuracy_scores) * 1.1])

    # Plot 2: Delta Accuracy and F1 in same plot
    delta_accuracy = non_baseline['delta_accuracy'].values * 100
    delta_f1 = non_baseline['delta_f1'].values

    x = np.arange(len(non_baseline))
    width = 0.35

    bars1 = ax2.bar(x - width/2, delta_accuracy, width, label='Δ Accuracy (%)',
                   color=COLORS['blue2'], alpha=0.8, edgecolor='black')
    bars2 = ax2.bar(x + width/2, delta_f1 * 100, width, label='Δ F1-Score (×100)',
                   color=COLORS['orange1'], alpha=0.8, edgecolor='black')

    ax2.axhline(y=0, color='black', linestyle='-', linewidth=1, alpha=0.5)
    ax2.set_ylabel('Delta Values', fontweight='bold', fontsize=12)
    ax2.set_title('B. Performance Impact vs Baseline (Accuracy & F1-Score)', fontweight='bold', fontsize=14)
    ax2.set_xticks(x)
    ax2.set_xticklabels(scenario_names, rotation=45, ha='right', fontsize=9)
    ax2.legend(fontsize=10, loc='upper right')
    ax2.grid(True, alpha=0.3, axis='y')

    # Set symmetric y-limits for delta plot
    max_delta = max(max(abs(delta_accuracy)), max(abs(delta_f1 * 100))) * 1.2
    ax2.set_ylim([-max_delta, max_delta])

    # Plot 3: Acceptance Rate comparison
    acceptance_rates = non_baseline['acceptance_rate'].values
    baseline_acceptance = baseline_result["acceptance_rate"]

    bars = ax3.bar(x_pos, acceptance_rates, color=COLORS['gray2'], alpha=0.8, edgecolor='black')
    ax3.axhline(y=baseline_acceptance, color=COLORS['gray3'], linestyle='--', linewidth=2,
                label=f'Baseline ({baseline_acceptance:.1f}%)')

    ax3.set_ylabel('Acceptance Rate (%)', fontweight='bold', fontsize=12)
    ax3.set_xlabel('Parameter Configurations', fontweight='bold', fontsize=12)
    ax3.set_title('C. Client Selection Rate', fontweight='bold', fontsize=14)
    ax3.set_xticks(x_pos)
    ax3.set_xticklabels(scenario_names, rotation=45, ha='right', fontsize=9)
    ax3.legend(fontsize=10, loc='upper right')
    ax3.grid(True, alpha=0.3, axis='y')
    ax3.set_ylim([0, 100])

    # Plot 4: Client impact (gained vs lost)
    gained = non_baseline['num_clients_gained'].values
    lost = non_baseline['num_clients_lost'].values

    width = 0.35
    bars1 = ax4.bar(x_pos - width/2, gained, width, label='Clients Gained',
                   color=COLORS['blue2'], alpha=0.8, edgecolor='black')
    bars2 = ax4.bar(x_pos + width/2, lost, width, label='Clients Lost',
                   color=COLORS['orange1'], alpha=0.8, edgecolor='black')

    ax4.set_ylabel('Number of Clients', fontweight='bold', fontsize=12)
    ax4.set_xlabel('Parameter Configurations', fontweight='bold', fontsize=12)
    ax4.set_title('D. Client Selection Impact vs Baseline', fontweight='bold', fontsize=14)
    ax4.set_xticks(x_pos)
    ax4.set_xticklabels(scenario_names, rotation=45, ha='right', fontsize=9)
    ax4.legend(fontsize=10, loc='upper right')
    ax4.grid(True, alpha=0.3, axis='y')

    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, 'q1_performance_comparison.png'), dpi=300, bbox_inches='tight')
    plt.savefig(os.path.join(FIG_DIR, 'q1_performance_comparison.pdf'), bbox_inches='tight')
    plt.show()

    return fig

def create_sensitivity_heatmap(results):
    """Fixed sensitivity heatmap for Q1 article"""
    print("\n🔥 Creating Sensitivity Analysis Heatmap...")

    df = pd.DataFrame(results)
    non_baseline = df[df['case_name'] != 'Baseline'].copy()

    if len(non_baseline) == 0:
        print("⚠️ No non-baseline results to plot")
        return None

    # Prepare data for heatmap - FIXED: Only multiply delta_accuracy by 100
    metrics = ['delta_accuracy', 'delta_f1', 'delta_roc_auc', 'delta_acceptance_rate', 'delta_pscore']
    metric_names = ['Δ Accuracy\n(%)', 'Δ F1-Score', 'Δ ROC-AUC', 'Δ Acceptance\nRate (%)', 'Δ P-score']

    heatmap_data = []
    for metric in metrics:
        if metric == 'delta_accuracy':
            values = non_baseline[metric].values * 100  # Convert to percentage
        elif metric == 'delta_acceptance_rate':
            values = non_baseline[metric].values  # Already in percentage
        else:
            values = non_baseline[metric].values
        heatmap_data.append(values)

    heatmap_data = np.array(heatmap_data)

    # Create professional heatmap
    fig, ax = plt.subplots(figsize=(14, 8))

    # Use diverging colormap for delta values with appropriate range
    vmax = max(2.0, np.abs(heatmap_data).max())
    im = ax.imshow(heatmap_data, cmap='RdBu_r', aspect='auto', vmin=-vmax, vmax=vmax)

    cbar = plt.colorbar(im, label='Delta Values', shrink=0.8)
    cbar.set_label('Delta Values', fontweight='bold', fontsize=12)

    ax.set_xlabel('Parameter Configurations', fontweight='bold', fontsize=12)
    ax.set_ylabel('Performance Metrics', fontweight='bold', fontsize=12)
    ax.set_title('Comprehensive Sensitivity Analysis: Performance Impact Matrix', fontweight='bold', fontsize=14)

    # Set labels
    case_names = [row['case_name'].replace('Weight_', 'W_').replace('Threshold_', 'T_').replace('Combined_', 'C_')
                 for _, row in non_baseline.iterrows()]

    ax.set_xticks(np.arange(len(non_baseline)))
    ax.set_xticklabels(case_names, rotation=45, ha='right', fontsize=9)
    ax.set_yticks(np.arange(len(metrics)))
    ax.set_yticklabels(metric_names, fontsize=10)

    # Add values to heatmap
    for i in range(len(metrics)):
        for j in range(len(non_baseline)):
            value = heatmap_data[i, j]

            if metrics[i] == 'delta_accuracy':
                text = f'{value:+.2f}%'
            elif metrics[i] == 'delta_acceptance_rate':
                text = f'{value:+.1f}%'
            else:
                text = f'{value:+.3f}'

            # Choose text color based on background
            text_color = 'white' if abs(value) > vmax/2 else 'black'

            ax.text(j, i, text, ha='center', va='center', fontsize=8, fontweight='bold', color=text_color)

    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, 'q1_sensitivity_heatmap.png'), dpi=300, bbox_inches='tight')
    plt.savefig(os.path.join(FIG_DIR, 'q1_sensitivity_heatmap.pdf'), bbox_inches='tight')
    plt.show()

    return fig

def create_parameter_sensitivity_figure(results, baseline_result):
    """Parameter sensitivity analysis figure"""
    print("\n📊 Creating Parameter Sensitivity Figure...")

    df = pd.DataFrame(results)

    # Filter out baseline for sensitivity analysis
    sensitivity_cases = df[df['case_name'] != 'Baseline'].copy()

    if len(sensitivity_cases) == 0:
        print("⚠️ No sensitivity cases to plot")
        return None

    # Group by parameter type
    weight_cases = sensitivity_cases[sensitivity_cases['case_type'] == 'weight_variation']
    threshold_cases = sensitivity_cases[sensitivity_cases['case_type'] == 'threshold_variation']
    combined_cases = sensitivity_cases[sensitivity_cases['case_type'] == 'combined_variation']

    # Create figure
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

    # Plot 1: Accuracy impact by parameter type
    categories = ['Weight Variations', 'Threshold Variations', 'Combined Variations']
    accuracy_impacts = [
        weight_cases['delta_accuracy'].mean() * 100 if len(weight_cases) > 0 else 0,
        threshold_cases['delta_accuracy'].mean() * 100 if len(threshold_cases) > 0 else 0,
        combined_cases['delta_accuracy'].mean() * 100 if len(combined_cases) > 0 else 0
    ]

    accuracy_stds = [
        weight_cases['delta_accuracy'].std() * 100 if len(weight_cases) > 0 else 0,
        threshold_cases['delta_accuracy'].std() * 100 if len(threshold_cases) > 0 else 0,
        combined_cases['delta_accuracy'].std() * 100 if len(combined_cases) > 0 else 0
    ]

    x_pos = np.arange(len(categories))
    bars = ax1.bar(x_pos, accuracy_impacts, yerr=accuracy_stds,
                   color=[COLORS['blue2'], COLORS['orange1'], COLORS['gray2']],
                   alpha=0.8, edgecolor='black', capsize=5)

    ax1.axhline(y=0, color='black', linestyle='-', linewidth=1, alpha=0.5)
    ax1.set_ylabel('Δ Accuracy (%)', fontweight='bold', fontsize=12)
    ax1.set_xlabel('Parameter Variation Type', fontweight='bold', fontsize=12)
    ax1.set_title('A. Average Accuracy Impact by Parameter Type', fontweight='bold', fontsize=14)
    ax1.set_xticks(x_pos)
    ax1.set_xticklabels(categories, fontsize=10)
    ax1.grid(True, alpha=0.3, axis='y')

    # Add value labels
    for bar, impact in zip(bars, accuracy_impacts):
        ax1.text(bar.get_x() + bar.get_width()/2.,
                impact + (0.1 if impact >= 0 else -0.3),
                f'{impact:+.2f}%',
                ha='center', va='bottom' if impact >= 0 else 'top',
                fontsize=10, fontweight='bold')

    # Plot 2: F1-Score impact by parameter type
    f1_impacts = [
        weight_cases['delta_f1'].mean() if len(weight_cases) > 0 else 0,
        threshold_cases['delta_f1'].mean() if len(threshold_cases) > 0 else 0,
        combined_cases['delta_f1'].mean() if len(combined_cases) > 0 else 0
    ]

    f1_stds = [
        weight_cases['delta_f1'].std() if len(weight_cases) > 0 else 0,
        threshold_cases['delta_f1'].std() if len(threshold_cases) > 0 else 0,
        combined_cases['delta_f1'].std() if len(combined_cases) > 0 else 0
    ]

    bars = ax2.bar(x_pos, f1_impacts, yerr=f1_stds,
                   color=[COLORS['blue2'], COLORS['orange1'], COLORS['gray2']],
                   alpha=0.8, edgecolor='black', capsize=5)

    ax2.axhline(y=0, color='black', linestyle='-', linewidth=1, alpha=0.5)
    ax2.set_ylabel('Δ F1-Score', fontweight='bold', fontsize=12)
    ax2.set_xlabel('Parameter Variation Type', fontweight='bold', fontsize=12)
    ax2.set_title('B. Average F1-Score Impact by Parameter Type', fontweight='bold', fontsize=14)
    ax2.set_xticks(x_pos)
    ax2.set_xticklabels(categories, fontsize=10)
    ax2.grid(True, alpha=0.3, axis='y')

    # Add value labels
    for bar, impact in zip(bars, f1_impacts):
        ax2.text(bar.get_x() + bar.get_width()/2.,
                impact + (0.001 if impact >= 0 else -0.003),
                f'{impact:+.4f}',
                ha='center', va='bottom' if impact >= 0 else 'top',
                fontsize=10, fontweight='bold')

    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, 'q1_parameter_sensitivity.png'), dpi=300, bbox_inches='tight')
    plt.savefig(os.path.join(FIG_DIR, 'q1_parameter_sensitivity.pdf'), bbox_inches='tight')
    plt.show()

    return fig

# ADD THIS DIAGNOSTIC FUNCTION
def debug_client_selection(client_ids, sensitivity_case):
    """Debug function to see why client selection isn't changing"""
    print(f"\n🔍 DEBUG Client Selection for {sensitivity_case['name']}:")

    dimension_scores = {cid: dims_for_client_realistic(cid, len(client_ids)) for cid in client_ids}
    scores = {cid: calculate_pscore(dimension_scores[cid], sensitivity_case["weights"]) for cid in client_ids}

    print(f"   P-scores: {scores}")
    print(f"   Score range: {min(scores.values()):.3f} to {max(scores.values()):.3f}")

    accepted, rejected = tadp_vr_partition(client_ids, scores,
                                         sensitivity_case["good_cut"],
                                         sensitivity_case["high_cut"],
                                         sensitivity_case["review_accept_ratio"])

    print(f"   Accepted: {accepted}")
    print(f"   Rejected: {rejected}")
    print(f"   Weights: {sensitivity_case['weights']}")

# MODIFY the main function to add debugging
def main():
    print("🚀 Starting TADP-VR Systematic Sensitivity Analysis")
    # ... existing code ...

    print("🔄 Preparing training data...")
    Xtr_all, Xte_global, ytr_all, yte_global, client_arrays = prepare_xy_without_reprocessing(raw_clients)
    print(f"✅ Data prepared: {Xtr_all.shape} train, {Xte_global.shape} test")

    # ADD: Debug client selection for a few key cases
    print("\n🔍 DEBUGGING CLIENT SELECTION:")
    debug_client_selection(client_ids, BASELINE_CASE)
    debug_client_selection(client_ids, SENSITIVITY_CASES[1])  # Weight_Balanced
    debug_client_selection(client_ids, SENSITIVITY_CASES[5])  # Weight_Extreme_Reliability

    # Continue with the rest...

def create_weight_analysis_figure(results):
    """Separate figure for weight analysis"""
    print("\n⚖️ Creating Weight Analysis Figure...")

    df = pd.DataFrame(results)
    weight_cases = df[df['case_type'] == 'weight_variation'].copy()

    if len(weight_cases) == 0:
        print("⚠️ No weight variation cases to plot")
        return None

    # Create figure
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

    # Plot 1: Weight distributions
    dimensions = ['dim1', 'dim2', 'dim3', 'dim4', 'dim5', 'dim6']
    dim_names = ['Reliability', 'Quality', 'Completeness', 'Consistency', 'Compliance', 'Timeliness']

    x = np.arange(len(dimensions))
    width = 0.12

    for i, (_, row) in enumerate(weight_cases.iterrows()):
        weights = [row['weights'][dim] for dim in dimensions]
        ax1.bar(x + i*width, weights, width, label=row['case_name'].replace('Weight_', ''),
                alpha=0.8, edgecolor='black')

    ax1.set_xlabel('Trust Dimensions', fontweight='bold', fontsize=12)
    ax1.set_ylabel('Weight Value', fontweight='bold', fontsize=12)
    ax1.set_title('A. Weight Distributions Across Scenarios', fontweight='bold', fontsize=14)
    ax1.set_xticks(x + width * (len(weight_cases)-1)/2)
    ax1.set_xticklabels(dim_names, rotation=45, ha='right')
    ax1.legend(fontsize=8, bbox_to_anchor=(1.05, 1), loc='upper left')
    ax1.grid(True, alpha=0.3, axis='y')

    # Plot 2: Performance vs Weight concentration
    weight_concentration = []
    performance_scores = []

    for _, row in weight_cases.iterrows():
        weights = list(row['weights'].values())
        # Calculate concentration (inverse of entropy)
        concentration = 1 / (-sum([w * np.log(w) for w in weights if w > 0]))
        weight_concentration.append(concentration)
        performance_scores.append(row['accuracy'] * 100)

    ax2.scatter(weight_concentration, performance_scores, s=100, alpha=0.7,
                color=COLORS['blue2'], edgecolor='black')

    # Add labels for each point
    for i, (_, row) in enumerate(weight_cases.iterrows()):
        ax2.annotate(row['case_name'].replace('Weight_', ''),
                    (weight_concentration[i], performance_scores[i]),
                    xytext=(5, 5), textcoords='offset points', fontsize=8)

    ax2.set_xlabel('Weight Concentration (Inverse Entropy)', fontweight='bold', fontsize=12)
    ax2.set_ylabel('Accuracy (%)', fontweight='bold', fontsize=12)
    ax2.set_title('B. Performance vs Weight Distribution Concentration', fontweight='bold', fontsize=14)
    ax2.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, 'q1_weight_analysis.png'), dpi=300, bbox_inches='tight')
    plt.savefig(os.path.join(FIG_DIR, 'q1_weight_analysis.pdf'), bbox_inches='tight')
    plt.show()

    return fig

# ---------------- Main Execution ----------------
def main():
    print("🚀 Starting TADP-VR Systematic Sensitivity Analysis")
    print("📥 Loading dataset...")

    global _DATASET_CACHE
    _DATASET_CACHE = None

    df = load_dataset()
    print(f"✅ Dataset loaded: {df.shape}")

    print("🔧 Preprocessing data...")
    df_processed = preprocess_diabetes(df)
    print(f"✅ Data preprocessed: {df_processed.shape}")

    print("👥 Splitting data into clients...")
    client_ids, raw_clients = split_clients(df_processed, num_clients=NUM_CLIENTS)
    print(f"✅ {len(client_ids)} clients created: {client_ids}")

    print("🔄 Preparing training data...")
    Xtr_all, Xte_global, ytr_all, yte_global, client_arrays = prepare_xy_without_reprocessing(raw_clients)
    print(f"✅ Data prepared: {Xtr_all.shape} train, {Xte_global.shape} test")

    print("\n" + "="*80)
    print("🔬 RUNNING SYSTEMATIC SENSITIVITY ANALYSIS")
    print("="*80)

    sensitivity_results, baseline_result = run_sensitivity_analysis(client_ids, raw_clients, client_arrays, Xte_global, yte_global)

    print("\n" + "="*80)
    print("📊 CREATING Q1 ARTICLE FIGURES AND TABLES")
    print("="*80)

    # Create professional Q1 article outputs
    config_table = create_configuration_table(sensitivity_results)
    impact_table = create_client_impact_table(sensitivity_results)
    weights_table = create_detailed_weights_table(sensitivity_results)
    summary_table = create_performance_summary_table(sensitivity_results)

    # Create separate figures for better readability
    performance_figure = create_performance_comparison_figure(sensitivity_results, baseline_result)
    heatmap_figure = create_sensitivity_heatmap(sensitivity_results)
    sensitivity_figure = create_parameter_sensitivity_figure(sensitivity_results, baseline_result)
    weight_figure = create_weight_analysis_figure(sensitivity_results)

    print("\n" + "="*80)
    print("✅ TADP-VR SYSTEMATIC SENSITIVITY ANALYSIS COMPLETED")
    print("="*80)

    # Print executive summary
    baseline_row = [r for r in sensitivity_results if r['case_name'] == 'Baseline'][0]

    print(f"\n📋 EXECUTIVE SUMMARY")
    print(f"   Baseline Performance: {baseline_row['accuracy']*100:.1f}% accuracy, {baseline_row['roc_auc']:.4f} ROC-AUC")
    print(f"   Baseline Acceptance: {baseline_row['acceptance_rate']:.1f}%")

    # Calculate impact ranges
    non_baseline = [r for r in sensitivity_results if r['case_name'] != 'Baseline']
    if non_baseline:
        accuracy_impacts = [r['delta_accuracy'] * 100 for r in non_baseline]
        f1_impacts = [r['delta_f1'] for r in non_baseline]
        roc_impacts = [r['delta_roc_auc'] for r in non_baseline]
        acceptance_impacts = [r['delta_acceptance_rate'] for r in non_baseline]

        print(f"   Accuracy Impact Range: {min(accuracy_impacts):+.1f}% to {max(accuracy_impacts):+.1f}%")
        print(f"   F1-Score Impact Range: {min(f1_impacts):+.4f} to {max(f1_impacts):+.4f}")
        print(f"   ROC-AUC Impact Range: {min(roc_impacts):+.4f} to {max(roc_impacts):+.4f}")
        print(f"   Acceptance Rate Impact Range: {min(acceptance_impacts):+.1f}% to {max(acceptance_impacts):+.1f}%")

        # Find most and least impactful scenarios
        most_accuracy = max(non_baseline, key=lambda x: x['delta_accuracy'])
        least_accuracy = min(non_baseline, key=lambda x: x['delta_accuracy'])
        most_f1 = max(non_baseline, key=lambda x: x['delta_f1'])
        least_f1 = min(non_baseline, key=lambda x: x['delta_f1'])

        print(f"\n🎯 KEY FINDINGS:")
        print(f"   Best Accuracy: {most_accuracy['case_name']} (+{most_accuracy['delta_accuracy']*100:.1f}%)")
        print(f"   Worst Accuracy: {least_accuracy['case_name']} ({least_accuracy['delta_accuracy']*100:+.1f}%)")
        print(f"   Best F1-Score: {most_f1['case_name']} (+{most_f1['delta_f1']:.4f})")
        print(f"   Worst F1-Score: {least_f1['case_name']} ({least_f1['delta_f1']:+.4f})")

        # Analysis by parameter type
        weight_cases = [r for r in non_baseline if r['case_type'] == 'weight_variation']
        threshold_cases = [r for r in non_baseline if r['case_type'] == 'threshold_variation']
        combined_cases = [r for r in non_baseline if r['case_type'] == 'combined_variation']

        print(f"\n📊 PARAMETER SENSITIVITY:")
        print(f"   Weight Variations: {len(weight_cases)} scenarios (avg ΔAcc: {np.mean([r['delta_accuracy']*100 for r in weight_cases]):+.2f}%)")
        print(f"   Threshold Variations: {len(threshold_cases)} scenarios (avg ΔAcc: {np.mean([r['delta_accuracy']*100 for r in threshold_cases]):+.2f}%)")
        print(f"   Combined Variations: {len(combined_cases)} scenarios (avg ΔAcc: {np.mean([r['delta_accuracy']*100 for r in combined_cases]):+.2f}%)")

if __name__ == "__main__":
    main()
```

```python
# ========================= TADP_VR_SYSTEMATIC_SENSITIVITY_ANALYSIS.py ===============
# Trustworthy AI Data Preparation (TADP) — Systematic Sensitivity Analysis
# ========================================================================

import os, io, glob, time, uuid, shutil, json, hashlib, warnings, math, random, threading, csv
from datetime import datetime, timezone
from typing import Dict, Any, List, Tuple

import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
import seaborn as sns
from IPython.display import display

warnings.filterwarnings("ignore")

# -------- Matplotlib appearance --------
mpl.rcParams.update({
    "figure.dpi": 600,
    "savefig.dpi": 600,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": True,
    "grid.alpha": 0.25,
    "legend.frameon": False,
    "font.size": 9,
    "figure.autolayout": True,
    "text.usetex": False,
    "font.family": "DejaVu Sans",
    "font.sans-serif": ["DejaVu Sans", "Liberation Sans"]
})

# Professional Q1 color scheme
COLORS = {
    'blue1': '#1f4e79',    # Dark blue
    'blue2': '#2e75b6',    # Medium blue
    'blue3': '#9dc3e6',    # Light blue
    'orange1': '#ed7d31',  # Orange
    'gray1': '#d0cece',    # Light gray
    'gray2': '#7f7f7f',    # Medium gray
    'gray3': '#2f2f2f',    # Dark gray
}

import psutil
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, MaxAbsScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score, accuracy_score, precision_score, recall_score, f1_score

# ---------------- Global config ----------------
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

# Configuration
RANDOM_STATE     = 42
N_RUNS           = 1
NUM_CLIENTS      = 15
NUM_ROUNDS_FL    = 4
LOCAL_EPOCHS     = 1
EPOCHS_CENTRAL   = 2
BATCH_SIZE       = 64
CLIENT_FRACTION  = 0.6

# TADP thresholds & P-score weights - BASELINE
GOOD_CUT = 3.0
HIGH_CUT = 3.5
WEIGHTS_PSCORE_DEFAULT = {"dim1":0.25,"dim2":0.15,"dim3":0.10,"dim4":0.10,"dim5":0.30,"dim6":0.10}

DATA_FILENAME  = "diabetes_130US.csv"
DATA_CACHE_DIR = "./data_cache"
os.makedirs(DATA_CACHE_DIR, exist_ok=True)

# File paths
TABLES_DIR = "./tables"
FIG_DIR    = os.path.join(TABLES_DIR, "figs")
for p in [TABLES_DIR, FIG_DIR]: os.makedirs(p, exist_ok=True)

# ---------------- Systematic Sensitivity Scenarios ----------------
def generate_weight_scenarios():
    """Generate 4 weight scenarios with sum = 1"""
    scenarios = []

    # Scenario 1: Balanced weights
    scenarios.append({
        "name": "Weight_Balanced",
        "weights": {"dim1":0.17,"dim2":0.17,"dim3":0.16,"dim4":0.16,"dim5":0.17,"dim6":0.17},
        "good_cut": GOOD_CUT,
        "high_cut": HIGH_CUT,
        "review_accept_ratio": 0.6,
        "type": "weight_variation"
    })

    # Scenario 2: Focus on Source Reliability and Data Quality
    scenarios.append({
        "name": "Weight_SourceData_Focus",
        "weights": {"dim1":0.30,"dim2":0.25,"dim3":0.10,"dim4":0.10,"dim5":0.15,"dim6":0.10},
        "good_cut": GOOD_CUT,
        "high_cut": HIGH_CUT,
        "review_accept_ratio": 0.6,
        "type": "weight_variation"
    })

    # Scenario 3: Focus on Regulatory Compliance
    scenarios.append({
        "name": "Weight_Regulatory_Focus",
        "weights": {"dim1":0.15,"dim2":0.15,"dim3":0.10,"dim4":0.10,"dim5":0.40,"dim6":0.10},
        "good_cut": GOOD_CUT,
        "high_cut": HIGH_CUT,
        "review_accept_ratio": 0.6,
        "type": "weight_variation"
    })

    # Scenario 4: Mixed emphasis
    scenarios.append({
        "name": "Weight_Mixed_Emphasis",
        "weights": {"dim1":0.20,"dim2":0.20,"dim3":0.15,"dim4":0.15,"dim5":0.20,"dim6":0.10},
        "good_cut": GOOD_CUT,
        "high_cut": HIGH_CUT,
        "review_accept_ratio": 0.6,
        "type": "weight_variation"
    })

    return scenarios

def generate_threshold_scenarios():
    """Generate 4 threshold scenarios"""
    scenarios = []

    # Scenario 1: Lenient thresholds
    scenarios.append({
        "name": "Threshold_Lenient",
        "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
        "good_cut": 2.5,
        "high_cut": 3.0,
        "review_accept_ratio": 0.6,
        "type": "threshold_variation"
    })

    # Scenario 2: Strict thresholds
    scenarios.append({
        "name": "Threshold_Strict",
        "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
        "good_cut": 3.2,
        "high_cut": 3.8,
        "review_accept_ratio": 0.6,
        "type": "threshold_variation"
    })

    # Scenario 3: Medium-High mix
    scenarios.append({
        "name": "Threshold_Medium_High",
        "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
        "good_cut": 2.8,
        "high_cut": 3.6,
        "review_accept_ratio": 0.6,
        "type": "threshold_variation"
    })

    # Scenario 4: High-High mix
    scenarios.append({
        "name": "Threshold_High_High",
        "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
        "good_cut": 3.3,
        "high_cut": 4.0,
        "review_accept_ratio": 0.6,
        "type": "threshold_variation"
    })

    return scenarios

def generate_combined_scenarios():
    """Generate 4 combined weight + threshold scenarios"""
    scenarios = []

    # Scenario 1: Balanced weights + Lenient thresholds
    scenarios.append({
        "name": "Combined_Balanced_Lenient",
        "weights": {"dim1":0.17,"dim2":0.17,"dim3":0.16,"dim4":0.16,"dim5":0.17,"dim6":0.17},
        "good_cut": 2.5,
        "high_cut": 3.0,
        "review_accept_ratio": 0.6,
        "type": "combined_variation"
    })

    # Scenario 2: SourceData focus + Strict thresholds
    scenarios.append({
        "name": "Combined_SourceData_Strict",
        "weights": {"dim1":0.30,"dim2":0.25,"dim3":0.10,"dim4":0.10,"dim5":0.15,"dim6":0.10},
        "good_cut": 3.2,
        "high_cut": 3.8,
        "review_accept_ratio": 0.6,
        "type": "combined_variation"
    })

    # Scenario 3: Regulatory focus + Medium thresholds
    scenarios.append({
        "name": "Combined_Regulatory_Medium",
        "weights": {"dim1":0.15,"dim2":0.15,"dim3":0.10,"dim4":0.10,"dim5":0.40,"dim6":0.10},
        "good_cut": 2.8,
        "high_cut": 3.4,
        "review_accept_ratio": 0.6,
        "type": "combined_variation"
    })

    # Scenario 4: Mixed emphasis + High thresholds
    scenarios.append({
        "name": "Combined_Mixed_High",
        "weights": {"dim1":0.20,"dim2":0.20,"dim3":0.15,"dim4":0.15,"dim5":0.20,"dim6":0.10},
        "good_cut": 3.3,
        "high_cut": 4.0,
        "review_accept_ratio": 0.6,
        "type": "combined_variation"
    })

    return scenarios

# Generate all sensitivity cases
BASELINE_CASE = {
    "name": "Baseline",
    "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
    "good_cut": GOOD_CUT,
    "high_cut": HIGH_CUT,
    "review_accept_ratio": 0.6,
    "type": "baseline"
}

# Add FedAvg case (true federated averaging with all clients)
FEDAVG_CASE = {
    "name": "FedAvg",
    "weights": WEIGHTS_PSCORE_DEFAULT.copy(),
    "good_cut": 0.0,  # Accept all clients
    "high_cut": 5.0,
    "review_accept_ratio": 1.0,
    "type": "fedavg"
}

SENSITIVITY_CASES = [BASELINE_CASE, FEDAVG_CASE] + generate_weight_scenarios() + generate_threshold_scenarios() + generate_combined_scenarios()

# ---------------- Core Functions ----------------
def preprocess_diabetes(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df.columns = df.columns.str.lower().str.replace(r'[^a-z0-9]+', '_', regex=True)
    df.replace('?', np.nan, inplace=True)

    if 'age' in df.columns and df['age'].dtype == object:
        df['age'] = df['age'].str.extract(r'(\d+)').astype(float)

    if 'readmitted' in df.columns:
        mapping = {'NO': 0, '<30': 1, '>30': 2, 'NONE': 0}
        s = df['readmitted'].astype(str).str.strip().str.upper()
        df['readmitted'] = s.map(mapping).fillna(0).astype('int8')

    df.drop(columns=['encounter_id', 'patient_nbr'], errors='ignore', inplace=True)
    return df

def build_preproc(df: pd.DataFrame, target: str = "readmitted"):
    from sklearn.preprocessing import StandardScaler
    from sklearn.compose import ColumnTransformer
    X = df.drop(columns=[target], errors="ignore").copy()
    numeric_cols = X.select_dtypes(include=['number']).columns.tolist()
    pre = ColumnTransformer(transformers=[("num", StandardScaler(), numeric_cols)], remainder="drop", sparse_threshold=0.0)
    return pre

def build_model(input_dim):
    inp = keras.Input(shape=(input_dim,), dtype=tf.float32)
    x = layers.Dense(128, activation="relu")(inp)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dropout(0.2)(x)
    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dropout(0.2)(x)
    x = layers.Dense(64, activation="relu")(x)
    out = layers.Dense(3, activation="softmax", dtype=tf.float32)(x)
    model = keras.Model(inp, out)
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=0.001), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    return model

def eval_metrics(model, Xte, yte):
    p = model.predict(Xte, verbose=0)
    yhat = np.argmax(p, axis=1)
    precision = precision_score(yte, yhat, average='macro', zero_division=0)
    recall = recall_score(yte, yhat, average='macro', zero_division=0)
    f1 = f1_score(yte, yhat, average='macro', zero_division=0)

    try:
        n_classes = len(np.unique(yte))
        if n_classes == 2:
            auc = roc_auc_score(yte, p[:, 1])
        elif n_classes == 3:
            try:
                auc = roc_auc_score(yte, p, multi_class='ovo', average='macro')
            except:
                auc = roc_auc_score(yte, p, multi_class='ovr', average='macro')
        else:
            auc = np.nan
    except:
        auc = np.nan

    return {"roc_auc": float(auc), "accuracy": float(accuracy_score(yte, yhat)), "precision": float(precision), "recall": float(recall), "f1": float(f1)}

def get_wvec(m):
    vec=[]
    for w in m.get_weights(): vec.append(w.flatten())
    return np.concatenate(vec) if vec else np.array([], np.float32)

def set_wvec(m, vec):
    shapes=[w.shape for w in m.get_weights()]
    out=[]; i=0
    for s in shapes:
        n=int(np.prod(s)); out.append(vec[i:i+n].reshape(s)); i+=n
    m.set_weights(out)

def split_clients(df: pd.DataFrame, num_clients=NUM_CLIENTS, target_col="readmitted", alpha=0.33):
    df = df.sample(frac=1, random_state=RANDOM_STATE).reset_index(drop=True)
    labels = df[target_col].values if target_col in df.columns else np.zeros(len(df), int)
    client_ids = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")[:num_clients]
    buckets = {cid: [] for cid in client_ids}
    rng = np.random.default_rng(RANDOM_STATE + 77)

    for y in np.unique(labels):
        idx = np.where(labels == y)[0]
        rng.shuffle(idx)
        props = rng.dirichlet([alpha] * num_clients)
        cuts = (np.cumsum(props) * len(idx)).astype(int)
        parts = np.split(idx, cuts[:-1])
        for cid, part in zip(client_ids, parts):
            buckets[cid].extend(part.tolist())

    client_data = {cid: df.loc[sorted(buckets[cid])].copy() for cid in client_ids}
    feature_num_cols = df.drop(columns=[target_col], errors="ignore").select_dtypes(include=np.number).columns.tolist()

    for j, cid in enumerate(client_ids):
        if not feature_num_cols: break
        shift = (j - (num_clients - 1) / 2.0) * 0.25
        noise = rng.normal(0, 0.05, size=len(feature_num_cols))
        client_data[cid].loc[:, feature_num_cols] = client_data[cid][feature_num_cols] + shift + noise

        if target_col in client_data[cid].columns:
            y = pd.to_numeric(client_data[cid][target_col], errors="coerce").fillna(0).astype(int)
            y = np.clip(y, 0, 2)
            client_data[cid][target_col] = y.astype(np.int8)

    return client_ids, client_data

def prepare_xy_without_reprocessing(clients: Dict[str, pd.DataFrame], target="readmitted", test_size=0.2):
    def can_stratify(y: np.ndarray, test_size: float) -> bool:
        vals, counts = np.unique(y, return_counts=True)
        if len(vals) < 2: return False
        for n in counts:
            if n < 2 or n * test_size < 1 or n * (1.0 - test_size) < 1: return False
        return True

    def sanitize_labels(y_like) -> np.ndarray:
        y = pd.to_numeric(pd.Series(y_like), errors="coerce").fillna(0).astype(int).values
        return np.clip(y, 0, 2).astype(np.int32)

    prepped = clients
    splits = {}
    train_parts, test_parts = [], []

    for cid, df in prepped.items():
        if target not in df.columns or len(df) == 0: continue
        df = df.copy()
        df[target] = sanitize_labels(df[target].values)
        X = df.drop(columns=[target])
        y = df[target].values
        strat = y if can_stratify(y, test_size) else None

        Xtr_raw, Xte_raw, ytr_raw, yte_raw = train_test_split(X, y, test_size=test_size, random_state=RANDOM_STATE, stratify=strat)
        ytr_raw = sanitize_labels(ytr_raw)
        yte_raw = sanitize_labels(yte_raw)
        splits[cid] = (Xtr_raw, Xte_raw, ytr_raw, yte_raw)
        train_parts.append(pd.concat([Xtr_raw.reset_index(drop=True), pd.Series(ytr_raw, name=target)], axis=1))
        test_parts.append(pd.concat([Xte_raw.reset_index(drop=True), pd.Series(yte_raw, name=target)], axis=1))

    if not train_parts: return np.array([]), np.array([]), np.array([]), np.array([]), {}

    train_all_df = pd.concat(train_parts, ignore_index=True)
    pre = build_preproc(train_all_df, target)
    pre.fit(train_all_df.drop(columns=[target], errors='ignore'))

    client_arrays = {}
    all_Xtr, all_ytr, all_Xte, all_yte = [], [], [], []
    for cid, (Xtr_raw, Xte_raw, ytr_raw, yte_raw) in splits.items():
        Xtr = pre.transform(Xtr_raw)
        Xte = pre.transform(Xte_raw)
        Xtr = Xtr.toarray() if hasattr(Xtr, "toarray") else np.asarray(Xtr, dtype=np.float32)
        Xte = Xte.toarray() if hasattr(Xte, "toarray") else np.asarray(Xte, dtype=np.float32)
        Xtr = np.nan_to_num(Xtr).astype(np.float32)
        Xte = np.nan_to_num(Xte).astype(np.float32)
        ytr = sanitize_labels(ytr_raw)
        yte = sanitize_labels(yte_raw)
        client_arrays[cid] = (Xtr, ytr)
        all_Xtr.append(Xtr); all_ytr.append(ytr)
        all_Xte.append(Xte); all_yte.append(yte)

    Xtr_all = np.vstack(all_Xtr) if all_Xtr else np.array([], dtype=np.float32)
    ytr_all = np.concatenate(all_ytr) if all_ytr else np.array([], dtype=np.int32)
    Xte_global = np.vstack(all_Xte) if all_Xte else np.array([], dtype=np.float32)
    yte_global = np.concatenate(all_yte) if all_yte else np.array([], dtype=np.int32)

    if yte_global.size and len(np.unique(yte_global)) < 2:
        all_df = pd.concat([df for df in prepped.values()], ignore_index=True)
        all_df = all_df.copy()
        all_df[target] = sanitize_labels(all_df[target].values)
        X_all = all_df.drop(columns=[target], errors='ignore')
        y_all = all_df[target].values
        strat_all = y_all if can_stratify(y_all, test_size) else None
        _, Xte_raw_fb, _, yte_raw_fb = train_test_split(X_all, y_all, test_size=test_size, random_state=RANDOM_STATE, stratify=strat_all)
        Xte_fb = pre.transform(Xte_raw_fb)
        Xte_fb = Xte_fb.toarray() if hasattr(Xte_fb, "toarray") else np.asarray(Xte_fb, dtype=np.float32)
        Xte_global = np.nan_to_num(Xte_fb).astype(np.float32)
        yte_global = sanitize_labels(yte_raw_fb)

    return Xtr_all, Xte_global, ytr_all, yte_global, client_arrays

def load_dataset():
    global _DATASET_CACHE
    cache_path = os.path.join(DATA_CACHE_DIR, DATA_FILENAME)
    if _DATASET_CACHE is not None: return _DATASET_CACHE
    if os.path.exists(cache_path):
        _DATASET_CACHE = pd.read_csv(cache_path); return _DATASET_CACHE
    if os.path.exists(DATA_FILENAME):
        df = pd.read_csv(DATA_FILENAME); df.to_csv(cache_path, index=False); _DATASET_CACHE = df; return df

    print("\n⚠️  Dataset not found in cache or working directory.")
    user_path = input("Paste local file path here (or press Enter to cancel): ").strip()
    if not user_path: raise FileNotFoundError("No dataset path provided. Aborting.")
    if not os.path.exists(user_path): raise FileNotFoundError(f"Provided path does not exist: {user_path}")
    shutil.copy2(user_path, cache_path)
    _DATASET_CACHE = pd.read_csv(cache_path)
    return _DATASET_CACHE

# ---------------- TADP-VR Core Functions ----------------
def dims_for_client_realistic(client_id: str, total_clients=NUM_CLIENTS):
    client_num = ord(client_id[0]) - ord('A') if client_id and client_id[0].isalpha() else 0
    client_position = client_num / max(1, total_clients - 1)

    if client_position < 0.2:
        base_scores = [2.2, 2.5, 2.8, 2.6, 2.4, 2.7]
    elif client_position < 0.8:
        middle_pos = (client_position - 0.2) / 0.6
        base_avg = 3.0 + (middle_pos * 0.5)
        base_scores = [base_avg - 0.2, base_avg - 0.1, base_avg, base_avg + 0.1, base_avg + 0.2, base_avg - 0.15]
    else:
        base_scores = [3.6, 3.8, 4.0, 3.7, 3.9, 3.8]

    random.seed(hash(client_id) % 1000)
    final_scores = []
    for base_score in base_scores:
        variation = random.uniform(-0.15, 0.15)
        final_score = max(0.0, min(5.0, base_score + variation))
        final_scores.append(round(final_score, 2))

    return {"dim1": final_scores[0], "dim2": final_scores[1], "dim3": final_scores[2],
            "dim4": final_scores[3], "dim5": final_scores[4], "dim6": final_scores[5]}

def calculate_pscore(dims: Dict[str,float], weights: Dict[str,float]):
    pscore = 0.0
    for dim, weight in weights.items():
        pscore += weight * dims[dim]
    return pscore

def tadp_vr_partition(client_ids: List[str], scores: Dict[str,float], good_cut, high_cut, review_accept_ratio=0.6):
    sorted_clients = sorted(client_ids, key=lambda cid: scores[cid])
    auto_reject = [cid for cid in sorted_clients if scores[cid] < good_cut]
    review = [cid for cid in sorted_clients if good_cut <= scores[cid] < high_cut]
    auto_accept = [cid for cid in sorted_clients if scores[cid] >= high_cut]

    review_sorted = sorted(review, key=lambda cid: scores[cid])
    review_accept_count = max(0, int(len(review) * review_accept_ratio))
    review_accept = review_sorted[-review_accept_count:] if review_sorted else []
    review_reject = review_sorted[:-review_accept_count] if review_sorted else []

    accepted = auto_accept + review_accept
    rejected = auto_reject + review_reject

    return accepted, rejected

def fed_round_sensitivity(global_model, round_idx, client_arrays, client_ids, sensitivity_case):
    import gc
    gc.collect()
    input_dim = global_model.input_shape[-1]

    dimension_scores = {cid: dims_for_client_realistic(cid, len(client_ids)) for cid in client_ids}
    scores = {cid: calculate_pscore(dimension_scores[cid], sensitivity_case["weights"]) for cid in client_ids}

    accepted, rejected = tadp_vr_partition(client_ids, scores, sensitivity_case["good_cut"], sensitivity_case["high_cut"], sensitivity_case["review_accept_ratio"])

    accepted = [cid for cid in accepted if cid in client_arrays]
    if not accepted:
        accepted = list(client_arrays.keys())[:2]
        print(f"    ⚠️  No clients accepted, using fallback: {accepted}")

    client_weights, client_n = [], []
    for cid in accepted:
        Xc, yc = client_arrays[cid]
        if len(Xc) == 0: continue
        cm = build_model(input_dim)
        set_wvec(cm, get_wvec(global_model))
        cm.fit(Xc, yc, epochs=LOCAL_EPOCHS, batch_size=BATCH_SIZE, verbose=0)
        client_weights.append(get_wvec(cm))
        client_n.append(len(Xc))

    if not client_weights:
        print("    ⚠️  No client updates received, keeping global model")
        return global_model, len(accepted), accepted, rejected, scores

    avg_weight = np.average(client_weights, axis=0, weights=client_n)
    set_wvec(global_model, avg_weight)

    return global_model, len(accepted), accepted, rejected, scores

def run_federated_sensitivity(client_ids, client_arrays, Xte, yte, sensitivity_case):
    print(f"\n🔬 Running Federated Sensitivity: {sensitivity_case['name']}")
    print(f"   Weights: {sensitivity_case['weights']}")
    print(f"   Thresholds: good_cut={sensitivity_case['good_cut']}, high_cut={sensitivity_case['high_cut']}")

    input_dim = Xte.shape[1]
    global_model = build_model(input_dim)

    start_time = time.time()
    total_clients_trained = 0
    round_clients = []
    all_accepted_clients = []
    all_rejected_clients = []
    all_scores = {}

    for round_idx in range(NUM_ROUNDS_FL):
        global_model, round_clients_count, accepted, rejected, scores = fed_round_sensitivity(
            global_model, round_idx, client_arrays, client_ids, sensitivity_case
        )
        total_clients_trained += round_clients_count
        round_clients.append(round_clients_count)
        all_accepted_clients = accepted
        all_rejected_clients = rejected
        all_scores = scores

    elapsed = time.time() - start_time
    metrics = eval_metrics(global_model, Xte, yte)
    acceptance_rate = len(all_accepted_clients) / len(client_ids) * 100

    return {
        "metrics": metrics,
        "elapsed": elapsed,
        "avg_clients_per_round": np.mean(round_clients),
        "total_clients_trained": total_clients_trained,
        "accepted_clients": all_accepted_clients,
        "rejected_clients": all_rejected_clients,
        "acceptance_rate": acceptance_rate,
        "client_scores": all_scores,
    }

def run_sensitivity_analysis(client_ids, raw_clients, client_arrays, Xte, yte):
    print("\n" + "="*80)
    print("🔬 TADP-VR SYSTEMATIC SENSITIVITY ANALYSIS")
    print("="*80)

    results = []
    baseline_result = None

    for case_idx, case in enumerate(SENSITIVITY_CASES):
        print(f"\n🧪 Sensitivity Case {case_idx+1}/{len(SENSITIVITY_CASES)}: {case['name']}")
        print(f"   Type: {case['type']}")
        print(f"   Weights: {case['weights']}")
        print(f"   Thresholds: good_cut={case['good_cut']}, high_cut={case['high_cut']}")

        try:
            case_result = run_federated_sensitivity(client_ids, client_arrays, Xte, yte, case)

            if case["name"] == "Baseline":
                baseline_result = case_result

            if baseline_result is not None and case["name"] not in ["Baseline", "FedAvg"]:
                delta_accuracy = case_result["metrics"]["accuracy"] - baseline_result["metrics"]["accuracy"]
                delta_f1 = case_result["metrics"]["f1"] - baseline_result["metrics"]["f1"]
                delta_roc_auc = case_result["metrics"]["roc_auc"] - baseline_result["metrics"]["roc_auc"]

                baseline_avg_pscore = np.mean([baseline_result["client_scores"][cid] for cid in baseline_result["accepted_clients"]]) if baseline_result["accepted_clients"] else 0
                case_avg_pscore = np.mean([case_result["client_scores"][cid] for cid in case_result["accepted_clients"]]) if case_result["accepted_clients"] else 0
                delta_pscore = case_avg_pscore - baseline_avg_pscore

                accepted_set = set(case_result["accepted_clients"])
                baseline_set = set(baseline_result["accepted_clients"])
                clients_gained = list(accepted_set - baseline_set)
                clients_lost = list(baseline_set - accepted_set)
                clients_kept = list(accepted_set & baseline_set)

                delta_acceptance_rate = case_result["acceptance_rate"] - baseline_result["acceptance_rate"]
            else:
                delta_accuracy = 0.0
                delta_f1 = 0.0
                delta_roc_auc = 0.0
                delta_pscore = 0.0
                delta_acceptance_rate = 0.0
                clients_gained = []
                clients_lost = []
                clients_kept = case_result["accepted_clients"]

            results.append({
                "case_name": case["name"],
                "case_type": case["type"],
                "weights": case["weights"],
                "good_cut": case["good_cut"],
                "high_cut": case["high_cut"],
                "accuracy": case_result["metrics"]["accuracy"],
                "f1": case_result["metrics"]["f1"],
                "roc_auc": case_result["metrics"]["roc_auc"],
                "precision": case_result["metrics"]["precision"],
                "recall": case_result["metrics"]["recall"],
                "acceptance_rate": case_result["acceptance_rate"],
                "avg_pscore": np.mean([case_result["client_scores"][cid] for cid in case_result["accepted_clients"]]) if case_result["accepted_clients"] else 0,
                "total_clients_trained": case_result["total_clients_trained"],
                "elapsed_time": case_result["elapsed"],
                "delta_accuracy": delta_accuracy,
                "delta_f1": delta_f1,
                "delta_roc_auc": delta_roc_auc,
                "delta_pscore": delta_pscore,
                "delta_acceptance_rate": delta_acceptance_rate,
                "clients_gained": clients_gained,
                "clients_lost": clients_lost,
                "num_clients_gained": len(clients_gained),
                "num_clients_lost": len(clients_lost),
            })

            print(f"✅ Completed: {case['name']}")

        except Exception as e:
            print(f"❌ Failed {case['name']}: {e}")
            import traceback
            traceback.print_exc()

    return results, baseline_result

# ---------------- REFINED TABLES FOR Q1 ARTICLE ----------------
def create_configuration_table(results):
    """Table 1: Scenario configuration and performance deltas"""
    print("\n📊 Creating Configuration and Performance Table...")

    df = pd.DataFrame(results)

    table_data = []
    for _, row in df.iterrows():
        weights_str = ", ".join([f"{k}:{v:.2f}" for k, v in row['weights'].items()])

        if row['case_type'] in ['baseline', 'fedavg']:
            table_data.append({
                'Scenario': row['case_name'],
                'Type': row['case_type'].title(),
                'good_cut': row['good_cut'],
                'high_cut': row['high_cut'],
                'Weights': weights_str,
                'Δ Accuracy': '0.00%',
                'Δ F1-Score': '0.0000',
                'Δ ROC-AUC': '0.0000',
                'Δ Acceptance Rate': '0.0%',
                'Acceptance Rate': f"{row['acceptance_rate']:.1f}%",
                'Time (s)': f"{row['elapsed_time']:.2f}"
            })
        else:
            table_data.append({
                'Scenario': row['case_name'],
                'Type': row['case_type'].replace('_', ' ').title(),
                'good_cut': row['good_cut'],
                'high_cut': row['high_cut'],
                'Weights': weights_str,
                'Δ Accuracy': f"{row['delta_accuracy']*100:+.2f}%",
                'Δ F1-Score': f"{row['delta_f1']:+.4f}",
                'Δ ROC-AUC': f"{row['delta_roc_auc']:+.4f}",
                'Δ Acceptance Rate': f"{row['delta_acceptance_rate']:+.1f}%",
                'Acceptance Rate': f"{row['acceptance_rate']:.1f}%",
                'Time (s)': f"{row['elapsed_time']:.2f}"
            })

    config_df = pd.DataFrame(table_data)

    print("\n" + "="*120)
    print("📋 TABLE 1: SENSITIVITY ANALYSIS CONFIGURATION AND PERFORMANCE")
    print("="*120)
    display(config_df)

    config_df.to_csv(os.path.join(TABLES_DIR, "sensitivity_configuration_table.csv"), index=False)
    print("✅ Configuration table saved to 'tables/sensitivity_configuration_table.csv'")

    return config_df

def create_client_impact_table(results):
    """Table 2: Client impact analysis"""
    print("\n📊 Creating Client Impact Table...")

    df = pd.DataFrame(results)

    table_data = []
    for _, row in df.iterrows():
        if row['case_type'] in ['baseline', 'fedavg']:
            table_data.append({
                'Scenario': row['case_name'],
                'Type': row['case_type'].title(),
                'Clients Gained': '-',
                'Clients Lost': '-',
                'Net Client Change': '0',
                'Acceptance Rate': f"{row['acceptance_rate']:.1f}%",
                'Avg P-score': f"{row['avg_pscore']:.3f}",
                'Δ P-score': '0.000',
                'Clients Trained': int(row['total_clients_trained'])
            })
        else:
            net_change = row['num_clients_gained'] - row['num_clients_lost']
            table_data.append({
                'Scenario': row['case_name'],
                'Type': row['case_type'].replace('_', ' ').title(),
                'Clients Gained': row['num_clients_gained'],
                'Clients Lost': row['num_clients_lost'],
                'Net Client Change': net_change,
                'Acceptance Rate': f"{row['acceptance_rate']:.1f}%",
                'Avg P-score': f"{row['avg_pscore']:.3f}",
                'Δ P-score': f"{row['delta_pscore']:+.3f}",
                'Clients Trained': int(row['total_clients_trained'])
            })

    impact_df = pd.DataFrame(table_data)

    print("\n" + "="*100)
    print("📋 TABLE 2: CLIENT IMPACT ANALYSIS")
    print("="*100)
    display(impact_df)

    impact_df.to_csv(os.path.join(TABLES_DIR, "client_impact_table.csv"), index=False)
    print("✅ Client impact table saved to 'tables/client_impact_table.csv'")

    return impact_df

# ---------------- Q1 ARTICLE FIGURES ----------------
def create_performance_comparison_figure(results, baseline_result):
    """Main performance comparison figure for Q1 article"""
    print("\n📈 Creating Performance Comparison Figure...")

    df = pd.DataFrame(results)
    non_baseline = df[~df['case_name'].isin(['Baseline', 'FedAvg'])].copy()

    if len(non_baseline) == 0:
        print("⚠️ No non-baseline results to plot")
        return None

    # Create professional figure
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

    # Get baseline and FedAvg values
    baseline_acc = baseline_result["metrics"]["accuracy"] * 100
    fedavg_row = df[df['case_name'] == 'FedAvg']
    fedavg_acc = fedavg_row['accuracy'].values[0] * 100 if len(fedavg_row) > 0 else baseline_acc

    # Plot 1: Accuracy comparison
    scenario_names = [row['case_name'].replace('Weight_', 'W_').replace('Threshold_', 'T_').replace('Combined_', 'C_')
                     for _, row in non_baseline.iterrows()]
    accuracy_scores = non_baseline['accuracy'].values * 100

    x_pos = np.arange(len(non_baseline))
    bars = ax1.bar(x_pos, accuracy_scores, color=COLORS['blue2'], alpha=0.8, edgecolor='black')

    # Add baseline and FedAvg lines
    ax1.axhline(y=baseline_acc, color=COLORS['blue1'], linestyle='--', linewidth=2,
                label=f'Baseline ({baseline_acc:.1f}%)')
    ax1.axhline(y=fedavg_acc, color=COLORS['orange1'], linestyle='--', linewidth=2,
                label=f'FedAvg ({fedavg_acc:.1f}%)')

    ax1.set_ylabel('Accuracy (%)', fontweight='bold', fontsize=12)
    ax1.set_title('A. Model Accuracy Across Parameter Configurations', fontweight='bold', fontsize=14)
    ax1.set_xticks(x_pos)
    ax1.set_xticklabels(scenario_names, rotation=45, ha='right', fontsize=9)
    ax1.legend(fontsize=10, loc='upper right')
    ax1.grid(True, alpha=0.3, axis='y')

    # Add value labels
    for bar, value in zip(bars, accuracy_scores):
        ax1.text(bar.get_x() + bar.get_width()/2., value + 0.5, f'{value:.1f}%',
                ha='center', va='bottom', fontsize=8, fontweight='bold')

    # Plot 2: Delta Accuracy impact
    delta_accuracy = non_baseline['delta_accuracy'].values * 100
    colors = [COLORS['blue2'] if x >= 0 else COLORS['orange1'] for x in delta_accuracy]

    bars = ax2.bar(x_pos, delta_accuracy, color=colors, alpha=0.8, edgecolor='black')
    ax2.axhline(y=0, color='black', linestyle='-', linewidth=1, alpha=0.5)
    ax2.set_ylabel('Δ Accuracy (%)', fontweight='bold', fontsize=12)
    ax2.set_title('B. Performance Impact vs Baseline', fontweight='bold', fontsize=14)
    ax2.set_xticks(x_pos)
    ax2.set_xticklabels(scenario_names, rotation=45, ha='right', fontsize=9)
    ax2.grid(True, alpha=0.3, axis='y')

    # Add value labels
    for bar, value in zip(bars, delta_accuracy):
        ax2.text(bar.get_x() + bar.get_width()/2.,
                value + (0.1 if value >= 0 else -0.3),
                f'{value:+.2f}%',
                ha='center', va='bottom' if value >= 0 else 'top',
                fontsize=8, fontweight='bold')

    # Plot 3: Acceptance Rate comparison
    acceptance_rates = non_baseline['acceptance_rate'].values
    baseline_acceptance = baseline_result["acceptance_rate"]

    bars = ax3.bar(x_pos, acceptance_rates, color=COLORS['gray2'], alpha=0.8, edgecolor='black')
    ax3.axhline(y=baseline_acceptance, color=COLORS['gray3'], linestyle='--', linewidth=2,
                label=f'Baseline ({baseline_acceptance:.1f}%)')
    ax3.axhline(y=100.0, color=COLORS['orange1'], linestyle='--', linewidth=2,
                label='FedAvg (100.0%)')

    ax3.set_ylabel('Acceptance Rate (%)', fontweight='bold', fontsize=12)
    ax3.set_xlabel('Parameter Configurations', fontweight='bold', fontsize=12)
    ax3.set_title('C. Client Selection Rate', fontweight='bold', fontsize=14)
    ax3.set_xticks(x_pos)
    ax3.set_xticklabels(scenario_names, rotation=45, ha='right', fontsize=9)
    ax3.legend(fontsize=10, loc='upper right')
    ax3.grid(True, alpha=0.3, axis='y')

    # Add value labels
    for bar, value in zip(bars, acceptance_rates):
        ax3.text(bar.get_x() + bar.get_width()/2., value + 0.5, f'{value:.1f}%',
                ha='center', va='bottom', fontsize=8, fontweight='bold')

    # Plot 4: Client impact (gained vs lost)
    gained = non_baseline['num_clients_gained'].values
    lost = non_baseline['num_clients_lost'].values

    width = 0.35
    bars1 = ax4.bar(x_pos - width/2, gained, width, label='Clients Gained',
                   color=COLORS['blue2'], alpha=0.8, edgecolor='black')
    bars2 = ax4.bar(x_pos + width/2, lost, width, label='Clients Lost',
                   color=COLORS['orange1'], alpha=0.8, edgecolor='black')

    ax4.set_ylabel('Number of Clients', fontweight='bold', fontsize=12)
    ax4.set_xlabel('Parameter Configurations', fontweight='bold', fontsize=12)
    ax4.set_title('D. Client Selection Impact vs Baseline', fontweight='bold', fontsize=14)
    ax4.set_xticks(x_pos)
    ax4.set_xticklabels(scenario_names, rotation=45, ha='right', fontsize=9)
    ax4.legend(fontsize=10, loc='upper right')
    ax4.grid(True, alpha=0.3, axis='y')

    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, 'q1_performance_comparison.png'), dpi=300, bbox_inches='tight')
    plt.savefig(os.path.join(FIG_DIR, 'q1_performance_comparison.pdf'), bbox_inches='tight')
    plt.show()

    return fig

def create_sensitivity_heatmap(results):
    """Comprehensive sensitivity heatmap for Q1 article"""
    print("\n🔥 Creating Sensitivity Analysis Heatmap...")

    df = pd.DataFrame(results)
    non_baseline = df[~df['case_name'].isin(['Baseline', 'FedAvg'])].copy()

    if len(non_baseline) == 0:
        print("⚠️ No non-baseline results to plot")
        return None

    # Prepare data for heatmap
    metrics = ['delta_accuracy', 'delta_f1', 'delta_roc_auc', 'delta_acceptance_rate', 'delta_pscore']
    metric_names = ['Δ Accuracy', 'Δ F1-Score', 'Δ ROC-AUC', 'Δ Acceptance', 'Δ P-score']

    heatmap_data = []
    for metric in metrics:
        if metric in ['delta_accuracy', 'delta_acceptance_rate']:
            values = non_baseline[metric].values * 100  # Convert to percentage
        else:
            values = non_baseline[metric].values
        heatmap_data.append(values)

    heatmap_data = np.array(heatmap_data)

    # Create professional heatmap
    fig, ax = plt.subplots(figsize=(14, 8))

    # Use diverging colormap for delta values
    im = ax.imshow(heatmap_data, cmap='RdBu_r', aspect='auto', vmin=-2, vmax=2)

    cbar = plt.colorbar(im, label='Delta Values', shrink=0.8)
    cbar.set_label('Delta Values', fontweight='bold', fontsize=12)

    ax.set_xlabel('Parameter Configurations', fontweight='bold', fontsize=12)
    ax.set_ylabel('Performance Metrics', fontweight='bold', fontsize=12)
    ax.set_title('Comprehensive Sensitivity Analysis: Performance Impact Matrix', fontweight='bold', fontsize=14)

    # Set labels
    case_names = [row['case_name'].replace('Weight_', 'W_').replace('Threshold_', 'T_').replace('Combined_', 'C_')
                 for _, row in non_baseline.iterrows()]

    ax.set_xticks(np.arange(len(non_baseline)))
    ax.set_xticklabels(case_names, rotation=45, ha='right', fontsize=9)
    ax.set_yticks(np.arange(len(metrics)))
    ax.set_yticklabels(metric_names, fontsize=10)

    # Add values to heatmap
    for i in range(len(metrics)):
        for j in range(len(non_baseline)):
            value = heatmap_data[i, j]

            if metrics[i] in ['delta_accuracy', 'delta_acceptance_rate']:
                text = f'{value:+.2f}%'
            else:
                text = f'{value:+.4f}'

            # Choose text color based on background
            text_color = 'white' if abs(value) > 1.0 else 'black'

            ax.text(j, i, text, ha='center', va='center', fontsize=8, fontweight='bold', color=text_color)

    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, 'q1_sensitivity_heatmap.png'), dpi=300, bbox_inches='tight')
    plt.savefig(os.path.join(FIG_DIR, 'q1_sensitivity_heatmap.pdf'), bbox_inches='tight')
    plt.show()

    return fig

def create_parameter_sensitivity_figure(results, baseline_result):
    """Parameter sensitivity analysis figure"""
    print("\n📊 Creating Parameter Sensitivity Figure...")

    df = pd.DataFrame(results)

    # Filter out baseline and FedAvg for sensitivity analysis
    sensitivity_cases = df[~df['case_name'].isin(['Baseline', 'FedAvg'])].copy()

    if len(sensitivity_cases) == 0:
        print("⚠️ No sensitivity cases to plot")
        return None

    # Group by parameter type
    weight_cases = sensitivity_cases[sensitivity_cases['case_type'] == 'weight_variation']
    threshold_cases = sensitivity_cases[sensitivity_cases['case_type'] == 'threshold_variation']
    combined_cases = sensitivity_cases[sensitivity_cases['case_type'] == 'combined_variation']

    # Create figure
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

    # Plot 1: Accuracy impact by parameter type
    categories = ['Weight Variations', 'Threshold Variations', 'Combined Variations']
    accuracy_impacts = [
        weight_cases['delta_accuracy'].mean() * 100 if len(weight_cases) > 0 else 0,
        threshold_cases['delta_accuracy'].mean() * 100 if len(threshold_cases) > 0 else 0,
        combined_cases['delta_accuracy'].mean() * 100 if len(combined_cases) > 0 else 0
    ]

    accuracy_stds = [
        weight_cases['delta_accuracy'].std() * 100 if len(weight_cases) > 0 else 0,
        threshold_cases['delta_accuracy'].std() * 100 if len(threshold_cases) > 0 else 0,
        combined_cases['delta_accuracy'].std() * 100 if len(combined_cases) > 0 else 0
    ]

    x_pos = np.arange(len(categories))
    bars = ax1.bar(x_pos, accuracy_impacts, yerr=accuracy_stds,
                   color=[COLORS['blue2'], COLORS['orange1'], COLORS['gray2']],
                   alpha=0.8, edgecolor='black', capsize=5)

    ax1.axhline(y=0, color='black', linestyle='-', linewidth=1, alpha=0.5)
    ax1.set_ylabel('Δ Accuracy (%)', fontweight='bold', fontsize=12)
    ax1.set_xlabel('Parameter Variation Type', fontweight='bold', fontsize=12)
    ax1.set_title('A. Average Accuracy Impact by Parameter Type', fontweight='bold', fontsize=14)
    ax1.set_xticks(x_pos)
    ax1.set_xticklabels(categories, fontsize=10)
    ax1.grid(True, alpha=0.3, axis='y')

    # Add value labels
    for bar, impact in zip(bars, accuracy_impacts):
        ax1.text(bar.get_x() + bar.get_width()/2.,
                impact + (0.1 if impact >= 0 else -0.3),
                f'{impact:+.2f}%',
                ha='center', va='bottom' if impact >= 0 else 'top',
                fontsize=10, fontweight='bold')

    # Plot 2: Acceptance rate impact by parameter type
    acceptance_impacts = [
        weight_cases['delta_acceptance_rate'].mean() if len(weight_cases) > 0 else 0,
        threshold_cases['delta_acceptance_rate'].mean() if len(threshold_cases) > 0 else 0,
        combined_cases['delta_acceptance_rate'].mean() if len(combined_cases) > 0 else 0
    ]

    acceptance_stds = [
        weight_cases['delta_acceptance_rate'].std() if len(weight_cases) > 0 else 0,
        threshold_cases['delta_acceptance_rate'].std() if len(threshold_cases) > 0 else 0,
        combined_cases['delta_acceptance_rate'].std() if len(combined_cases) > 0 else 0
    ]

    bars = ax2.bar(x_pos, acceptance_impacts, yerr=acceptance_stds,
                   color=[COLORS['blue2'], COLORS['orange1'], COLORS['gray2']],
                   alpha=0.8, edgecolor='black', capsize=5)

    ax2.axhline(y=0, color='black', linestyle='-', linewidth=1, alpha=0.5)
    ax2.set_ylabel('Δ Acceptance Rate (%)', fontweight='bold', fontsize=12)
    ax2.set_xlabel('Parameter Variation Type', fontweight='bold', fontsize=12)
    ax2.set_title('B. Average Client Selection Impact by Parameter Type', fontweight='bold', fontsize=14)
    ax2.set_xticks(x_pos)
    ax2.set_xticklabels(categories, fontsize=10)
    ax2.grid(True, alpha=0.3, axis='y')

    # Add value labels
    for bar, impact in zip(bars, acceptance_impacts):
        ax2.text(bar.get_x() + bar.get_width()/2.,
                impact + (0.5 if impact >= 0 else -1.0),
                f'{impact:+.1f}%',
                ha='center', va='bottom' if impact >= 0 else 'top',
                fontsize=10, fontweight='bold')

    plt.tight_layout()
    plt.savefig(os.path.join(FIG_DIR, 'q1_parameter_sensitivity.png'), dpi=300, bbox_inches='tight')
    plt.savefig(os.path.join(FIG_DIR, 'q1_parameter_sensitivity.pdf'), bbox_inches='tight')
    plt.show()

    return fig

# ---------------- Main Execution ----------------
def main():
    print("🚀 Starting TADP-VR Systematic Sensitivity Analysis")
    print("📥 Loading dataset...")

    global _DATASET_CACHE
    _DATASET_CACHE = None

    df = load_dataset()
    print(f"✅ Dataset loaded: {df.shape}")

    print("🔧 Preprocessing data...")
    df_processed = preprocess_diabetes(df)
    print(f"✅ Data preprocessed: {df_processed.shape}")

    print("👥 Splitting data into clients...")
    client_ids, raw_clients = split_clients(df_processed, num_clients=NUM_CLIENTS)
    print(f"✅ {len(client_ids)} clients created: {client_ids}")

    print("🔄 Preparing training data...")
    Xtr_all, Xte_global, ytr_all, yte_global, client_arrays = prepare_xy_without_reprocessing(raw_clients)
    print(f"✅ Data prepared: {Xtr_all.shape} train, {Xte_global.shape} test")

    print("\n" + "="*80)
    print("🔬 RUNNING SYSTEMATIC SENSITIVITY ANALYSIS")
    print("="*80)

    sensitivity_results, baseline_result = run_sensitivity_analysis(client_ids, raw_clients, client_arrays, Xte_global, yte_global)

    print("\n" + "="*80)
    print("📊 CREATING Q1 ARTICLE FIGURES AND TABLES")
    print("="*80)

    # Create professional Q1 article outputs
    config_table = create_configuration_table(sensitivity_results)
    impact_table = create_client_impact_table(sensitivity_results)
    performance_figure = create_performance_comparison_figure(sensitivity_results, baseline_result)
    heatmap_figure = create_sensitivity_heatmap(sensitivity_results)
    sensitivity_figure = create_parameter_sensitivity_figure(sensitivity_results, baseline_result)

    print("\n" + "="*80)
    print("✅ TADP-VR SYSTEMATIC SENSITIVITY ANALYSIS COMPLETED")
    print("="*80)

    # Print executive summary
    baseline_row = [r for r in sensitivity_results if r['case_name'] == 'Baseline'][0]
    fedavg_row = [r for r in sensitivity_results if r['case_name'] == 'FedAvg'][0]

    print(f"\n📋 EXECUTIVE SUMMARY")
    print(f"   Baseline Performance: {baseline_row['accuracy']*100:.1f}% accuracy, {baseline_row['roc_auc']:.4f} ROC-AUC")
    print(f"   FedAvg Performance: {fedavg_row['accuracy']*100:.1f}% accuracy, {fedavg_row['roc_auc']:.4f} ROC-AUC")
    print(f"   Baseline Acceptance: {baseline_row['acceptance_rate']:.1f}%")
    print(f"   FedAvg Acceptance: {fedavg_row['acceptance_rate']:.1f}%")

    # Calculate impact ranges
    non_baseline = [r for r in sensitivity_results if r['case_name'] not in ['Baseline', 'FedAvg']]
    if non_baseline:
        accuracy_impact = max([r['delta_accuracy'] for r in non_baseline]) * 100
        roc_impact = max([r['delta_roc_auc'] for r in non_baseline])

        print(f"   Maximum Accuracy Impact: {accuracy_impact:+.1f}%")
        print(f"   Maximum ROC-AUC Impact: {roc_impact:+.4f}")

        # Find most and least impactful scenarios
        most_accuracy = max(non_baseline, key=lambda x: x['delta_accuracy'])
        least_accuracy = min(non_baseline, key=lambda x: x['delta_accuracy'])

        print(f"\n🎯 KEY FINDINGS:")
        print(f"   Most Positive Impact: {most_accuracy['case_name']} (+{most_accuracy['delta_accuracy']*100:.1f}% accuracy)")
        print(f"   Most Negative Impact: {least_accuracy['case_name']} ({least_accuracy['delta_accuracy']*100:+.1f}% accuracy)")

if __name__ == "__main__":
    main()
```
